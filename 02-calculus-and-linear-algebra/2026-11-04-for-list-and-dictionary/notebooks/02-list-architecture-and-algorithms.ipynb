# Notebook 02: Python List Architecture, Mutations, and Algorithmic Operations

This notebook explores the internal structure, operations, and algorithmic patterns of Python lists:
1. **Dynamic Array Architecture:** Indexing, negative indexing, and slicing `[start:stop:step]`.
2. **Mutations & In-Place Methods:** `append()`, `insert()`, `extend()`, `pop()`, `remove()`, and `clear()`.
3. **Sorting & Reversal:** In-place `sort()` / `reverse()` vs non-mutating `sorted()` / `reversed()`.
4. **Algorithmic Aggregations:** Summing even numbers, summing items at even indices, and running stats.


```python
# Memory representation and indexing
data = [4, 5, 6, 7, 8, 9]

# Forward indexing: 0 to len - 1
assert data[0] == 4
assert data[3] == 7

# Backward indexing: -1 to -len
assert data[-1] == 9
assert data[-3] == 7

# Slicing semantics: list[start:end:step]
assert data[2:4] == [6, 7]
assert data[:3] == [4, 5, 6]
assert data[3:] == [7, 8, 9]
assert data[::2] == [4, 6, 8]
assert data[::-1] == [9, 8, 7, 6, 5, 4]

print("Indexing and slicing assertions verified!")
```

## 1. List Mutation Mechanics

Contrasting $O(1)$ amortized appends vs $O(n)$ arbitrary insertions and deletions.


```python
items = [6, 5, 7, 1, 9, 2]

# Append: O(1) amortized
items.append(4)
assert items == [6, 5, 7, 1, 9, 2, 4]

# Insert at index 0: O(n) element shift
items.insert(0, 8)
assert items == [8, 6, 5, 7, 1, 9, 2, 4]

# Extend: batch append
items.extend([10, 11])
assert items[-2:] == [10, 11]

# Pop by index: O(1) from end, O(n) from middle
last = items.pop()
assert last == 11
first = items.pop(0)
assert first == 8

# Remove by value: removes first matching instance
items.remove(5)
assert 5 not in items

print("List mutation assertions verified!")
```

## 2. In-Place vs. Pure Functional Operations

Demonstrating the difference between mutating a list in-place and returning a new sorted/reversed container.


```python
raw_data = [6, 5, 7, 1, 9, 2]

# sorted() returns a brand new list, leaving raw_data untouched
asc = sorted(raw_data)
desc = sorted(raw_data, reverse=True)
assert raw_data == [6, 5, 7, 1, 9, 2]
assert asc == [1, 2, 5, 6, 7, 9]
assert desc == [9, 7, 6, 5, 2, 1]

# .sort() mutates in place and returns None
copy_data = raw_data.copy()
ret = copy_data.sort()
assert ret is None
assert copy_data == [1, 2, 5, 6, 7, 9]

# Reversal: reversed() iterator vs .reverse() in place
rev_iter = list(reversed(raw_data))
assert rev_iter == [2, 9, 1, 7, 5, 6]
copy_data.reverse()
assert copy_data == [9, 7, 6, 5, 2, 1]

print("Sorting and reversal assertions verified!")
```

## 3. Algorithmic Aggregations: Value Filtering vs. Index Filtering

We compare two common interview and data-cleaning patterns:
- **Sum of even values:** Filtering on element magnitude (`x % 2 == 0`).
- **Sum of even index values:** Filtering on positional index (`idx % 2 == 0`).


```python
def sum_even_values(nums: list[int]) -> int:
    """Compute sum of all even integers in nums."""
    total = 0
    for x in nums:
        if x % 2 == 0:
            total += x
    return total


def sum_even_indices(nums: list[int]) -> int:
    """Compute sum of all elements situated at even positions (0, 2, 4, ...)."""
    total = 0
    for idx in range(len(nums)):
        if idx % 2 == 0:
            total += nums[idx]
    return total


sample = [6, 5, 7, 1, 9, 2]

# Even values in sample: 6 + 2 = 8
even_val_sum = sum_even_values(sample)
assert even_val_sum == 8

# Elements at even indices: sample[0]=6, sample[2]=7, sample[4]=9 -> 6 + 7 + 9 = 22
even_idx_sum = sum_even_indices(sample)
assert even_idx_sum == 22

print(f"Sample: {sample}")
print(f"Sum of even values: {even_val_sum}")
print(f"Sum of even index elements: {even_idx_sum}")
print("All list algorithms verified successfully!")
```
