## Goal

Load grayscale and color images from `../data/nature.jpg` using Matplotlib and OpenCV,
inspect their `(Height, Width)` and `(Height, Width, Channel)` tensor shapes, and
convert seamlessly between NumPy arrays and pure-Python 2D lists.

```python
from pathlib import Path

import cv2
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import numpy as np

DATA_DIR = Path("../data")
OUTPUT_DIR = Path("../outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
IMAGE_PATH = DATA_DIR / "nature.jpg"
```

## 1. Loading and Displaying with Matplotlib and OpenCV

```python
# Load RGB with Matplotlib
img_rgb = mpimg.imread(IMAGE_PATH)
print("Matplotlib RGB shape:", img_rgb.shape, "dtype:", img_rgb.dtype)

# Load grayscale with OpenCV (flag 0 == cv2.IMREAD_GRAYSCALE)
img_gray = cv2.imread(str(IMAGE_PATH), cv2.IMREAD_GRAYSCALE)
print("OpenCV Grayscale shape:", img_gray.shape, "dtype:", img_gray.dtype)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(img_rgb)
axes[0].set_title("Color Image (H, W, 3)")
axes[0].axis("off")

axes[1].imshow(img_gray, cmap="gray")
axes[1].set_title("Grayscale Image (H, W)")
axes[1].axis("off")

plt.tight_layout()
plt.show()

cv2.imwrite(str(OUTPUT_DIR / "01_nature_gray.jpg"), img_gray)
```

## 2. Using a Pure-Python 2D List as an Intermediate Representation

```python
# Convert NumPy array -> 2D Python list
gray_list = img_gray.tolist()
height = len(gray_list)
width = len(gray_list[0])
print(f"2D list dimensions -> Height: {height}, Width: {width}")

# Convert 2D Python list -> uint8 NumPy array and save
reconstructed = np.asarray(gray_list, dtype=np.uint8)
assert np.array_equal(reconstructed, img_gray)
cv2.imwrite(str(OUTPUT_DIR / "01_nature_roundtrip.jpg"), reconstructed)
```
