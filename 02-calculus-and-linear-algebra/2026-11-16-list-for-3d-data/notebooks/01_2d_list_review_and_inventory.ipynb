## 1. Main Diagonal Sum (`sum_main_diagonal`)

For a square matrix $A \in \mathbb{R}^{n \times n}$, the main diagonal runs from top-left $(0, 0)$ to bottom-right $(n - 1, n - 1)$:

$$\operatorname{tr}(A) = \sum_{i=0}^{n-1} A_{i, i}$$

Always validate that the matrix is non-empty and that every row has length equal to the number of rows before indexing `matrix[i][i]`.

```python
def sum_main_diagonal(matrix: list[list[float]]) -> float:
    """Return the sum of main-diagonal elements of a non-empty square matrix."""
    if not matrix or not matrix[0]:
        raise ValueError("Matrix must be non-empty.")

    num_rows = len(matrix)
    if any(len(row) != num_rows for row in matrix):
        raise ValueError("Matrix is not square.")

    diagonal_sum = 0.0
    for i in range(num_rows):
        diagonal_sum += matrix[i][i]
    return diagonal_sum


sample_matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
trace_value = sum_main_diagonal(sample_matrix)
print("Main diagonal sum:", trace_value)
assert trace_value == 15

try:
    sum_main_diagonal([[1, 2, 3], [4, 5, 6]])
except ValueError as exc:
    print("Validation check passed:", exc)
```

## 2. Sorting Matrix Rows by Row Sum (`sort_by_row_sum`)

Compare two standard Python patterns for sorting rows in ascending order of their element sum:

1. **Explicit `(row_sum, row)` pairs:** pair each row with its sum, sort explicitly by `pair[0]` so ties preserve original row order (Timsort stability), and unpack the sorted rows.
2. **Direct `key=sum`:** pass `sum` directly to `sorted(matrix, key=sum)`.

```python
def sort_by_row_sum_pairs(matrix: list[list[float]]) -> list[list[float]]:
    """Sort rows by ascending row sum using explicit (sum, row) pairs."""
    row_sums = [(sum(row), row) for row in matrix]
    sorted_pairs = sorted(row_sums, key=lambda pair: pair[0])
    return [row for _, row in sorted_pairs]


def sort_by_row_sum_key(matrix: list[list[float]]) -> list[list[float]]:
    """Sort rows by ascending row sum using Python's built-in key=sum."""
    return sorted(matrix, key=sum)


matrix_to_sort = [
    [4, 2, 7],  # sum = 13
    [1, 5, 6],  # sum = 12
    [3, 8, 2],  # sum = 13
]

sorted_pairs_result = sort_by_row_sum_pairs(matrix_to_sort)
sorted_key_result = sort_by_row_sum_key(matrix_to_sort)

print("Sorted via (sum, row) pairs:")
for row in sorted_pairs_result:
    print(" ", row, "-> sum =", sum(row))

assert sorted_pairs_result == sorted_key_result
assert sorted_pairs_result == [[1, 5, 6], [4, 2, 7], [3, 8, 2]]
```

## 3. Inventory Management Matrix ($n \times 3$)

An $n \times 3$ nested list can store heterogeneous tabular records where each row holds `[product_name, quantity, unit_price]`:

- Compute total inventory quantity across all products.
- Compute per-product inventory value (`quantity * unit_price`).
- Compute total inventory valuation across the catalog.

```python
inventory = [
    ["Product 1", 10, 5.0],
    ["Product 2", 5, 2.5],
    ["Product 3", 3, 8.0],
]

total_quantity = sum(item[1] for item in inventory)
product_values = [item[1] * item[2] for item in inventory]
total_value = sum(product_values)

print("Total inventory quantity:", total_quantity)
print("Per-product values:", product_values)
print("Total inventory value:", total_value)

assert total_quantity == 18
assert product_values == [50.0, 12.5, 24.0]
assert total_value == 86.5
```
