## Goal

Use `GridSearchCV` to exhaustively evaluate a discrete Cartesian hyperparameter grid
for a Support Vector Classifier (`SVC`) across linear and RBF kernels on a synthetic
50-feature classification dataset.

> **Trade-off:** Grid Search guarantees evaluating every point in the specified discrete
> grid, making it straightforward and reproducible, but its runtime grows exponentially
> with the number of tuned hyperparameters.

```python
import time

from sklearn import model_selection, svm
from sklearn.datasets import make_classification
```

## 1. Generate Synthetic Classification Dataset

```python
X, y = make_classification(
    n_samples=300,
    n_features=50,
    n_informative=10,
    n_redundant=25,
    n_repeated=15,
    n_clusters_per_class=5,
    flip_y=0.05,
    class_sep=0.5,
    random_state=0,
)
print(f"Dataset shape: X={X.shape}, y={y.shape}")
```

## 2. Define Conditional Grid & Run 5-Fold GridSearchCV

```python
svc = svm.SVC(probability=True, random_state=1)

# Conditional grids: 'gamma' applies only to the RBF kernel
search_grid = [
    {"C": [1, 10, 100, 1000], "kernel": ["linear"]},
    {"C": [1, 10, 100, 1000], "gamma": [0.001, 0.0001], "kernel": ["rbf"]},
]

search_func = model_selection.GridSearchCV(
    estimator=svc,
    param_grid=search_grid,
    scoring="accuracy",
    n_jobs=-1,
    cv=5,
)

start = time.perf_counter()
search_func.fit(X, y)
elapsed = time.perf_counter() - start

print(f"GridSearchCV finished in {elapsed:.2f}s")
print("Best params:", search_func.best_params_)
print(f"Best 5-fold CV accuracy: {search_func.best_score_:.4f}")
```
