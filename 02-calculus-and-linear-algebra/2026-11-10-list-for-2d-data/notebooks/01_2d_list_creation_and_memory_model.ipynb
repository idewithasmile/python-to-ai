## Goal

Understand how Python stores 2D lists in memory, diagnose the classic shallow-multiplication
bug (`[[0] * W] * H`), compare shallow vs. deep copying, and enforce rectangular shape invariants.

## 1. The Shallow-Multiplication Trap vs. List Comprehension

```python
rows, cols = 3, 4

# Anti-pattern: multiplies references to the exact same inner list
aliased_grid = [[0] * cols] * rows
print("Row object IDs (aliased):", [id(row) for row in aliased_grid])

aliased_grid[0][2] = 99
print("After mutating aliased_grid[0][2] = 99:")
for row in aliased_grid:
    print(" ", row)
```

```python
# Safe pattern: list comprehension allocates an independent list per row
independent_grid = [[0] * cols for _ in range(rows)]
print("Row object IDs (independent):", [id(row) for row in independent_grid])

independent_grid[0][2] = 99
print("After mutating independent_grid[0][2] = 99:")
for row in independent_grid:
    print(" ", row)
```

## 2. Shallow Copy vs. Deep Copy of 2D Matrices

```python
import copy

original = [[1, 2, 3], [4, 5, 6]]

shallow_copy = original.copy()  # only copies the outer list of references
row_slice_copy = [row[:] for row in original]  # independent rows for scalar entries
deep_copy = copy.deepcopy(original)

shallow_copy[0][0] = -100
print("Original after mutating shallow_copy:", original)

row_slice_copy[1][2] = 777
deep_copy[1][2] = 888
print("Original after mutating row_slice_copy and deep_copy:", original)
print("row_slice_copy:", row_slice_copy)
print("deep_copy:", deep_copy)
```

## 3. Rectangular Shape Validation

```python
def matrix_shape(matrix: list[list[float]]) -> tuple[int, int]:
    """Return (rows, cols) for a non-empty rectangular 2D list."""
    if not matrix or not matrix[0]:
        raise ValueError("Matrix must be non-empty.")
    num_cols = len(matrix[0])
    if any(len(row) != num_cols for row in matrix):
        raise ValueError("All rows must have the same length.")
    return len(matrix), num_cols


print("Valid 2x3 shape:", matrix_shape([[1, 2, 3], [4, 5, 6]]))

try:
    matrix_shape([[1, 2], [3]])
except ValueError as exc:
    print("Caught jagged matrix:", exc)
```
