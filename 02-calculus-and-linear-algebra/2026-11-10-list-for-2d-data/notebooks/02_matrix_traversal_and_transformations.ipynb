## Goal

Implement core 2D list traversals and transformations—column slicing, main/anti-diagonal
sums, matrix transpose, 90-degree rotation, and row-aggregate sorting—verified against
the reference fixtures in `../data/matrix_benchmarks.json`.

```python
import json
from pathlib import Path

with Path("../data/matrix_benchmarks.json").open(encoding="utf-8") as fp:
    benchmarks = json.load(fp)

square = benchmarks["square_matrix"]
unsorted = benchmarks["unsorted_rows_matrix"]
print("Square matrix:", square)
```

## 1. Diagonals and Column Extraction

```python
def diagonal_sums(matrix: list[list[int]]) -> tuple[int, int]:
    """Compute (main_diagonal_sum, anti_diagonal_sum) of a square matrix."""
    n = len(matrix)
    if n == 0 or any(len(row) != n for row in matrix):
        raise ValueError("Matrix must be non-empty and square.")
    main_sum = sum(matrix[i][i] for i in range(n))
    anti_sum = sum(matrix[i][n - 1 - i] for i in range(n))
    return main_sum, anti_sum


def get_column(matrix: list[list[int]], col_idx: int) -> list[int]:
    """Extract column *col_idx* from a 2D list."""
    return [row[col_idx] for row in matrix]


main_d, anti_d = diagonal_sums(square)
assert main_d == benchmarks["expected_main_diagonal_sum"]
assert anti_d == benchmarks["expected_anti_diagonal_sum"]
print(f"Main diagonal sum: {main_d}, Anti-diagonal sum: {anti_d}")
print("Column 1:", get_column(square, 1))
```

## 2. Transpose, 90-Degree Rotation, and Row-Sum Sorting

```python
def transpose(matrix: list[list[int]]) -> list[list[int]]:
    """Return the transpose of a rectangular 2D list."""
    rows = len(matrix)
    cols = len(matrix[0])
    return [[matrix[i][j] for i in range(rows)] for j in range(cols)]


def rotate_90_clockwise(matrix: list[list[int]]) -> list[list[int]]:
    """Rotate a matrix 90 degrees clockwise via transpose + row reversal."""
    return [row[::-1] for row in transpose(matrix)]


def sort_rows_by_sum(matrix: list[list[int]]) -> list[list[int]]:
    """Return a new matrix with rows ordered by non-decreasing row sum."""
    return [row[:] for row in sorted(matrix, key=sum)]


print("Transpose:", transpose([[1, 2, 3], [4, 5, 6]]))
print("90 deg CW:", rotate_90_clockwise(square))

sorted_rows = sort_rows_by_sum(unsorted)
assert sorted_rows == benchmarks["expected_sorted_by_row_sum"]
print("Sorted by row sum:", sorted_rows)
```
