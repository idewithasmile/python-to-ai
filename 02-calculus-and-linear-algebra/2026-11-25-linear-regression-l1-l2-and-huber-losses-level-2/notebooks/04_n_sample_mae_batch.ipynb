# Notebook 04 — Full-Batch Gradient Descent with MAE

## Goal

Implement full-batch gradient descent with **MAE**.  Accumulate sign-based gradients over
all $N$ samples before updating weights.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## 1  Load Data

```python
def get_column(data: list, index: int) -> list:
    return [row[index] for row in data]


data = np.genfromtxt("../data/data.csv", delimiter=",").tolist()
x_data = get_column(data, 0)
y_data = get_column(data, 1)
N = len(x_data)
```

## 2  Model Functions

```python
def predict(x: float, w: float, b: float) -> float:
    return w * x + b


def gradient_mae(y_hat: float, y: float, x: float) -> tuple[float, float]:
    residual = y_hat - y
    dw = x * residual / abs(residual)
    db = residual / abs(residual)
    return dw, db


def update_weight(w: float, b: float, lr: float, dw: float, db: float) -> tuple[float, float]:
    return w - lr * dw, b - lr * db
```

## 3  Batch Gradient Descent Loop

```python
w: float = -0.34
b: float = 0.04
lr: float = 0.01
epoch_max: int = 60

epoch_losses: list[float] = []

for _epoch in range(epoch_max):
    dw_total = 0.0
    db_total = 0.0
    loss_total = 0.0

    for i in range(N):
        x = x_data[i]
        y = y_data[i]

        y_hat = predict(x, w, b)
        loss_total += abs(y_hat - y)

        dw, db = gradient_mae(y_hat, y, x)
        dw_total += dw
        db_total += db

    epoch_losses.append(loss_total / N)
    w, b = update_weight(w, b, lr, dw_total / N, db_total / N)

print(f"Final parameters  →  w = {w:.4f}, b = {b:.4f}")
```

## 4  Loss Curve and Fitted Line

```python
plt.figure(figsize=(6, 4))
plt.plot(epoch_losses)
plt.xlabel("Epoch")
plt.ylabel("Mean MAE loss")
plt.title("Full-Batch GD — MAE Loss per Epoch")
plt.tight_layout()
plt.show()
```

```python
x_line = list(range(2, 8))
y_line = [predict(x, w, b) for x in x_line]

plt.figure(figsize=(6, 4))
plt.scatter(x_data, y_data, label="Observations")
plt.plot(x_line, y_line, "r-", label=f"Fit: y = {w:.3f}x + {b:.3f}")
plt.xlabel("Area (×100 m²)")
plt.ylabel("Price (Tael)")
plt.title("Linear Regression — MAE (Batch GD)")
plt.xlim(3, 7)
plt.ylim(4, 10)
plt.legend()
plt.tight_layout()
plt.show()
```
