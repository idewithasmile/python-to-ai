## Goal

Visualize the 1D Gaussian probability density function, examine how taking the natural
logarithm converts bell-shaped exponential decay into a concave quadratic curve, and
compare a positive parabolic function with its logarithmic transform.

```python
import math

import matplotlib.pyplot as plt


def gaussian(x: float, mean: float, variance: float) -> float:
    """Evaluate the 1D Gaussian PDF at *x*."""
    part1 = 1.0 / math.sqrt(2.0 * math.pi * variance)
    part2 = math.exp(-(1.0 / (2.0 * variance)) * (x - mean) ** 2)
    return part1 * part2
```

## 1. Coarse vs. Fine Grid 1D Gaussian PDF

```python
mean = 10.0
variance = 4.0

coarse_x = list(range(-10, 31))
coarse_y = [gaussian(x, mean, variance) for x in coarse_x]

k = 5
fine_x = [x / k for x in range(-10 * k, 30 * k + 1)]
fine_y = [gaussian(x, mean, variance) for x in fine_x]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(coarse_x, coarse_y, marker="o", markersize=3)
axes[0].set_title("1D Gaussian (Integer Grid)")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")

axes[1].plot(fine_x, fine_y)
axes[1].set_title("1D Gaussian (Step 0.2)")
axes[1].set_xlabel("x")
axes[1].set_ylabel("f(x)")

plt.tight_layout()
plt.show()
```

## 2. Log-Gaussian and Log-Parabolic Transforms

```python
log_gaussian_y = [math.log(val) for val in fine_y]

k_parabola = 10
x_parabola = [x / k_parabola for x in range(-5 * k_parabola, 5 * k_parabola + 1)]
squared_y = [x**2 + 1e-3 for x in x_parabola]
log_squared_y = [math.log(val) for val in squared_y]

fig, axes = plt.subplots(1, 3, figsize=(13, 4))
axes[0].plot(fine_x, log_gaussian_y)
axes[0].set_title("Log-Gaussian: Concave Quadratic")
axes[0].set_xlabel("x")
axes[0].set_ylabel("ln f(x)")

axes[1].plot(x_parabola, squared_y)
axes[1].set_title("Parabolic Function: x^2 + 1e-3")
axes[1].set_xlabel("x")
axes[1].set_ylabel("g(x)")

axes[2].plot(x_parabola, log_squared_y)
axes[2].set_title("Log of Parabolic Function")
axes[2].set_xlabel("x")
axes[2].set_ylabel("ln g(x)")

plt.tight_layout()
plt.show()
```
