# Two Pointers: Convergent Pointers and In-Place Reversals

**Session:** 2026-11-02 · Recursion and Two Pointer  
**Module:** 01 · Python Fundamentals & Algorithmic Thinking

The **Two Pointers** technique is an algorithmic pattern where two indices traverse a data structure (such as an array or string) either towards each other (**convergent / opposite-ends**) or in the same direction (**sliding window / fast-slow**).

Core advantages:
- Reduces time complexity from $O(n^2)$ pairwise comparisons to a single $O(n)$ linear scan.
- Achieves optimal $O(1)$ auxiliary space by performing in-place mutations.


## 1. Indexing Mechanics in Python Lists

Python supports dual indexing systems:
- **Positive Indexing:** $0, 1, \dots, n-1$ indexing elements from left to right.
- **Negative Indexing:** $-n, -(n-1), \dots, -1$ indexing elements from right to left.
- Symmetry: `arr[i]` corresponds to `arr[i - n]`, so `arr[0] == arr[-n]` and `arr[n-1] == arr[-1]`.


```python
arr = [1, 2, 1, 5, 3]
n = len(arr)

print(f"First element: arr[0]={arr[0]}, arr[-{n}]={arr[-n]}")
print(f"Second element: arr[1]={arr[1]}, arr[-{n - 1}]={arr[-n + 1]}")
print(f"Last element: arr[{n - 1}]={arr[n - 1]}, arr[-1]={arr[-1]}")

assert arr[0] == arr[-5]
assert arr[1] == arr[-4]
assert arr[4] == arr[-1]
```

## 2. In-Place String and Array Reversal (LeetCode 344)

Given a string or list of characters, reverse it in-place.
- A naive loop scans backwards and creates a new string: $O(n)$ time, $O(n)$ space.
- Convergent Two Pointers: Initialize `left = 0` and `right = n - 1`. While `left < right`, swap `s[left]` and `s[right]`, then advance `left += 1` and decrement `right -= 1`.


```python
def reverse_string_for_loop(s: str) -> str:
    """Reverse string using a backwards for loop.

    Time Complexity: O(n)
    Auxiliary Space: O(n)
    """
    result = []
    for i in range(len(s) - 1, -1, -1):
        result.append(s[i])
    return "".join(result)


def reverse_string_two_pointers(s: str) -> str:
    """Reverse string using two pointers on a mutable list of characters.

    Time Complexity: O(n)
    Auxiliary Space: O(n) for list conversion (O(1) in-place on list)
    """
    chars = list(s)
    i = 0
    j = len(chars) - 1
    while i < j:
        chars[i], chars[j] = chars[j], chars[i]
        i += 1
        j -= 1
    return "".join(chars)


# Tests from lecture slides
for test_str in ["hello", "Hannah"]:
    res_loop = reverse_string_for_loop(test_str)
    res_two = reverse_string_two_pointers(test_str)
    print(f"Original: {test_str!r} -> Reversed: {res_two!r}")
    assert res_loop == res_two
```

## 3. Palindrome Verification (LeetCode 125)

A string is a palindrome if it reads the same forward and backward. In practical applications (text processing, DNA sequencing), we often verify palindromes while ignoring non-alphanumeric characters and casing.


```python
def is_palindrome_strict(s: str) -> bool:
    """Strict two-pointer palindrome check."""
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True


def is_palindrome_alphanumeric(s: str) -> bool:
    """In-place two-pointer palindrome check ignoring punctuation and case.

    Time Complexity: O(n)
    Auxiliary Space: O(1)
    """
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True


assert is_palindrome_strict("racecar") is True
assert is_palindrome_strict("Hannah") is False  # casing mismatch
assert is_palindrome_alphanumeric("Hannah") is True
assert is_palindrome_alphanumeric("A man, a plan, a canal: Panama") is True
assert is_palindrome_alphanumeric("algorithmic") is False
print("All palindrome tests passed!")
```

## 4. Two Sum on a Sorted Array (LeetCode 167)

Given a 1-indexed array of integers that is already **sorted in non-decreasing order**, find two numbers such that they add up to a specific `target` number.

**Mathematical Invariant:**
- Let `current_sum = arr[left] + arr[right]`.
- If `current_sum == target`: solution found.
- If `current_sum < target`: because the array is sorted, pairing `arr[left]` with any element smaller than `arr[right]` will yield an even smaller sum. Thus `arr[left]` cannot be part of the solution with any remaining candidate. We must increment `left += 1`.
- If `current_sum > target`: symmetrically, `arr[right]` is too large to pair with any element greater than `arr[left]`. We must decrement `right -= 1`.


```python
from collections.abc import Sequence


def two_sum_sorted(nums: Sequence[int], target: int) -> tuple[int, int] | None:
    """Find 0-based index pair (left, right) summing to target in sorted array.

    Time Complexity: O(n)
    Auxiliary Space Complexity: O(1)
    """
    left = 0
    right = len(nums) - 1
    while left < right:
        curr = nums[left] + nums[right]
        if curr == target:
            return left, right
        elif curr < target:
            left += 1
        else:
            right -= 1
    return None


# Test against benchmark
sample_sorted = [2, 7, 11, 15]
idx = two_sum_sorted(sample_sorted, 9)
print("Two sum target 9 on [2, 7, 11, 15]:", idx)
assert idx == (0, 1)

idx_none = two_sum_sorted(sample_sorted, 100)
assert idx_none is None
```
