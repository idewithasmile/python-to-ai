## 1. Fast and Slow Pointers: Deduplication in Sorted Arrays (LeetCode 26)

Given a sorted array `nums`, remove duplicates in-place such that each unique element appears only once.
- `slow` tracks the index of the last placed unique element.
- `fast` scans ahead from index $1$ to $n-1$.
- Whenever `nums[fast] != nums[slow]`, advance `slow += 1` and write `nums[slow] = nums[fast]`.


```python
def remove_duplicates_sorted(nums: list[int]) -> int:
    """Remove duplicates in-place from sorted array using slow-fast pointers.

    Time Complexity: O(n)
    Auxiliary Space: O(1)
    Returns: Number of unique elements k.
    """
    if not nums:
        return 0
    slow = 0
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]
    return slow + 1


arr_dup = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = remove_duplicates_sorted(arr_dup)
print(f"Unique count: {k}, modified prefix: {arr_dup[:k]}")
assert k == 5
assert arr_dup[:k] == [0, 1, 2, 3, 4]
```

## 2. Multi-Pointer Search: 3Sum (LeetCode 15)

Given an integer array `nums`, find all unique triplets $[a, b, c]$ such that $a + b + c = 0$.
- Brute Force: $O(n^3)$ cubic triple loop.
- Two Pointers with Sorting:
  1. Sort array in $O(n \log n)$.
  2. Iterate $i$ from $0$ to $n-3$. If $i > 0$ and `nums[i] == nums[i-1]`, skip to avoid duplicates.
  3. Set `left = i + 1`, `right = n - 1`, and run convergent two pointers to find pairs summing to `-nums[i]`.
  4. Total Time Complexity: $O(n^2)$, Auxiliary Space: $O(1)$.


```python
def three_sum(nums: list[int]) -> list[tuple[int, int, int]]:
    """Find all unique triplets summing to zero.

    Time Complexity: O(n^2)
    Auxiliary Space: O(1) auxiliary (excluding output)
    """
    nums_sorted = sorted(nums)
    triplets = []
    n = len(nums_sorted)

    for i in range(n - 2):
        if i > 0 and nums_sorted[i] == nums_sorted[i - 1]:
            continue
        left = i + 1
        right = n - 1
        target = -nums_sorted[i]

        while left < right:
            current_sum = nums_sorted[left] + nums_sorted[right]
            if current_sum == target:
                triplets.append((nums_sorted[i], nums_sorted[left], nums_sorted[right]))
                while left < right and nums_sorted[left] == nums_sorted[left + 1]:
                    left += 1
                while left < right and nums_sorted[right] == nums_sorted[right - 1]:
                    right -= 1
                left += 1
                right -= 1
            elif current_sum < target:
                left += 1
            else:
                right -= 1
    return triplets


sample_3sum = [-1, 0, 1, 2, -1, -4]
res_3sum = three_sum(sample_3sum)
print("3Sum triplets:", res_3sum)
assert (-1, -1, 2) in res_3sum
assert (-1, 0, 1) in res_3sum
assert len(res_3sum) == 2
```

## 3. 3Sum Closest (LeetCode 16)

Given an integer array `nums` and a `target`, find three integers such that the sum is closest to `target`. Return the sum of the three integers.


```python
def three_sum_closest(nums: list[int], target: int) -> int:
    """Find the triplet sum closest to target.

    Time Complexity: O(n^2)
    Auxiliary Space: O(1)
    """
    nums_sorted = sorted(nums)
    n = len(nums_sorted)
    closest_sum = nums_sorted[0] + nums_sorted[1] + nums_sorted[2]

    for i in range(n - 2):
        left = i + 1
        right = n - 1
        while left < right:
            curr = nums_sorted[i] + nums_sorted[left] + nums_sorted[right]
            if abs(curr - target) < abs(closest_sum - target):
                closest_sum = curr
            if curr < target:
                left += 1
            elif curr > target:
                right -= 1
            else:
                return target
    return closest_sum


nums_closest = [-1, 2, 1, -4]
res_closest = three_sum_closest(nums_closest, 1)
print("3Sum closest to 1 in [-1, 2, 1, -4]:", res_closest)
assert res_closest == 2  # -1 + 2 + 1 = 2
```

## 4. Matrix Transposition and 2D Indexing (Slide 63)

The transpose of matrix $A \in \mathbb{R}^{m \times n}$ is matrix $A^T \in \mathbb{R}^{n \times m}$ defined by:
$$A^T_{i, j} = A_{j, i}$$

For a square matrix ($n \times n$), transposition can be executed in-place by swapping elements across the main diagonal: `matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]` for all $j > i$.


```python
def transpose_matrix(matrix: list[list[int]]) -> list[list[int]]:
    """Transpose rectangular or square matrix."""
    rows = len(matrix)
    cols = len(matrix[0])
    return [[matrix[r][c] for r in range(rows)] for c in range(cols)]


def transpose_square_in_place(matrix: list[list[int]]) -> None:
    """Transpose square matrix in-place via diagonal reflection.

    Time Complexity: O(n^2)
    Auxiliary Space: O(1)
    """
    n = len(matrix)
    for i in range(n):
        for j in range(i + 1, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]


mat = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
transpose_square_in_place(mat)
print("Transposed matrix:", mat)
assert mat == [[1, 4, 7], [2, 5, 8], [3, 6, 9]]
```
