In classification problems, measuring model performance solely through accuracy is dangerous, especially in imbalanced regimes (e.g., medical diagnostics, fraud detection, anomaly alerting). An algorithm that predicts 'negative' 100% of the time on a dataset with 99% negative cases achieves 99% accuracy while having zero clinical or operational value.

To accurately evaluate performance, machine learning engineers use **Precision**, **Recall**, and their harmonic mean, the **F1-Score**.

### Objectives
1. Understand the 2x2 binary classification Confusion Matrix ($TP, FP, FN, TN$).
2. Formulate Precision and Recall and explore the Precision-Recall trade-off.
3. Understand why the harmonic mean is mathematically preferred over the arithmetic mean.
4. Implement a robust evaluation function with strict type checking and defensive bounds.


## 1. The Confusion Matrix and Metric Formulations

Consider a binary classification scenario:
- **True Positive ($TP$):** Ground truth is positive, model predicts positive.
- **False Positive ($FP$):** Ground truth is negative, model predicts positive (Type I error, false alarm).
- **False Negative ($FN$):** Ground truth is positive, model predicts negative (Type II error, missed detection).
- **True Negative ($TN$):** Ground truth is negative, model predicts negative.

```text
                     Actual Positive      Actual Negative
Predicted Positive   True Positive (TP)   False Positive (FP)   -> Precision = TP / (TP + FP)
Predicted Negative   False Negative (FN)  True Negative (TN)
                     v
             Recall = TP / (TP + FN)
```

### Mathematical Definitions

1. **Precision:**
   $$\text{Precision} = \frac{TP}{TP + FP} = \frac{\text{Correct Positive Detections}}{\text{All Positive Detections}}$$

2. **Recall (Sensitivity / True Positive Rate):**
   $$\text{Recall} = \frac{TP}{TP + FN} = \frac{\text{Correct Positive Detections}}{\text{All Ground-Truth Positives}}$$

3. **F1-Score:**
   The harmonic mean of Precision and Recall:
   $$F_1 = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}} = \frac{2 \cdot TP}{2 \cdot TP + FP + FN}$$


## 2. Why Harmonic Mean Over Arithmetic Mean?

Suppose a model predicts positive for every single instance:
- Recall = $1.0$ (it found all positives).
- Precision = $0.05$ (only 5% of predictions were actually positive).

If we use the **arithmetic mean**:
$$\frac{1.0 + 0.05}{2} = 0.525 \quad (52.5\%)$$
This misleadingly suggests a mediocre model.

With the **harmonic mean**:
$$F_1 = 2 \cdot \frac{1.0 \cdot 0.05}{1.0 + 0.05} = \frac{0.10}{1.05} \approx 0.0952 \quad (9.5\%)$$
The harmonic mean is strictly bounded by the smaller component, properly penalizing severe imbalance.


## 3. Implementing the Evaluation Function

According to the specification:
1. Inputs: `tp`, `fp`, `fn`.
2. Type validation:
   - Check if `tp` is an `int`. If not, print `'tp must be int'` and terminate.
   - Check if `fp` is an `int`. If not, print `'fp must be int'` and terminate.
   - Check if `fn` is an `int`. If not, print `'fn must be int'` and terminate.
3. Boundary validation:
   - `tp`, `fp`, and `fn` must all be strictly greater than zero ($> 0$).
   - If not, print `'tp and fp and fn must be greater than zero'` and terminate.
4. Calculate and print Precision, Recall, and F1-Score.


````python
def calculate_f1_score(tp, fp, fn) -> tuple[float, float, float] | None:
    """Evaluates classification performance with precision, recall, and f1-score.

    Includes strict type enforcement and boundary validation as per lab specifications.
    """
    # Explicit type verification (excluding bool, which is a subclass of int in Python)
    if type(tp) is not int:
        print("tp must be int")
        return None
    if type(fp) is not int:
        print("fp must be int")
        return None
    if type(fn) is not int:
        print("fn must be int")
        return None

    # Boundary verification: strictly greater than 0
    if not (tp > 0 and fp > 0 and fn > 0):
        print("tp and fp and fn must be greater than zero")
        return None

    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    f1_score = 2 * (precision * recall) / (precision + recall)

    print(f"precision is {precision}")
    print(f"recall is {recall}")
    print(f"f1-score is {f1_score}")

    return precision, recall, f1_score
````

## 4. Validating Correct Execution and Error Handlers

We now test our implementation against standard valid cases and error scenarios.


````python
# Valid input case
print("=== Valid Case: tp=50, fp=10, fn=20 ===")
res = calculate_f1_score(tp=50, fp=10, fn=20)

# Invalid type cases
print("\n=== Error Case: tp is float ===")
calculate_f1_score(tp=2.5, fp=10, fn=20)

print("\n=== Error Case: fn is string ===")
calculate_f1_score(tp=50, fp=10, fn="twenty")

# Out-of-bounds cases (<= 0)
print("\n=== Error Case: zero value ===")
calculate_f1_score(tp=0, fp=10, fn=20)

print("\n=== Error Case: negative value ===")
calculate_f1_score(tp=50, fp=-5, fn=20)
````

## 5. Verification Assertions


````python
import math

# Analytical verification
prec, rec, f1 = calculate_f1_score(50, 10, 20)
expected_prec = 50 / 60
expected_rec = 50 / 70
expected_f1 = 2 * (expected_prec * expected_rec) / (expected_prec + expected_rec)

assert math.isclose(prec, expected_prec)
assert math.isclose(rec, expected_rec)
assert math.isclose(f1, expected_f1)
print("Classification metric tests passed successfully!")
````
