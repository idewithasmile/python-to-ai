## 1. Parameters of `print()`

The signature of `print()` is:
```python
print(*objects, sep=' ', end='\n', file=None, flush=False)
```
- `*objects`: Zero or more positional arguments. Python calls `str()` on each object to obtain its textual representation.
- `sep`: String inserted between objects. Defaults to a single space `' '`.
- `end`: String appended after the last object. Defaults to a newline `'\n'`.
- `file`: Stream object supporting `.write(str)`. Defaults to `sys.stdout`.
- `flush`: Boolean controlling whether the output buffer is immediately flushed to the underlying operating system file descriptor.


````python
import sys

# Basic printing with default parameters
print("Epoch", 1, "Loss", 0.4231)

# Custom delimiters using sep
print("2026", "10", "24", sep="-")
print("usr", "local", "bin", sep="/")
print("tensor", "shape", "is", "3x224x224", sep=" -> ")
print("A", "B", "C", sep="")
````

## 2. Line Termination with `end`

By default, every `print()` call appends `'\n'`. Overriding `end` enables inline logging, multi-stage progress reporting, and custom formatting without newline overhead.


````python
# Inline iteration with space delimiter
print("Processing batches: ", end="")
for batch_idx in range(5):
    print(f"[{batch_idx}]", end=" ")
print("Complete!")

# Multi-segment header separator
print("Metric", end=" | ")
print("Train Score", end=" | ")
print("Val Score", end="\n")
print("-" * 35)
````

## 3. Output Streams and Buffering (`file` and `flush`)

Standard streams are split into:
- `sys.stdout`: Standard output for routine program output. Typically line-buffered when connected to a terminal, or block-buffered when piped.
- `sys.stderr`: Standard error stream for diagnostics, warnings, and error messages. Typically unbuffered so errors appear immediately even during unexpected crashes.

The `flush=True` keyword argument forces Python to flush the write buffer immediately, critical for real-time progress bars or interactive command loops.


````python
import io

# Directing errors to stderr
print("WARNING: Gradient norm exceeded threshold; clipping applied.", file=sys.stderr)

# Directing output to an in-memory string stream
buffer = io.StringIO()
print("Model Summary", file=buffer)
print("Layers: 12, Parameters: 85M", file=buffer)

captured_text = buffer.getvalue()
print("Captured from buffer:")
print(captured_text.strip())
````

## 4. Modern Formatted String Literals (f-strings)

Introduced in Python 3.6 (PEP 498), f-strings provide fast runtime expression evaluation and format specifications.

| Format Specifier | Meaning | Example |
| :--- | :--- | :--- |
| `:.4f` | Fixed-point float with 4 decimals | `f"{0.123456:.4f}"` -> `"0.1235"` |
| `:05d` | Integer zero-padded to width 5 | `f"{42:05d}"` -> `"00042"` |
| `:.2%` | Percentage format with 2 decimals | `f"{0.875:.2%}"` -> `"87.50%"` |
| `:.2e` | Scientific exponential notation | `f"{1e-4:.2e}"` -> `"1.00e-04"` |
| `:<10` / `:>10` | Left / Right align within width 10 | `f"{'loss':>10}"` |


````python
learning_rate = 0.00035
accuracy = 0.94827
step = 420
model_name = "TransformerEncoder"

# Numerical formatting
print(f"Step [{step:06d}] | LR: {learning_rate:.2e} | Acc: {accuracy:.2%}")

# Expression evaluation and function execution inside f-strings
print(f"Model: {model_name.upper()} | Name Length: {len(model_name)}")

# Self-documenting debugging specifier (=)
batch_size = 64
dropout = 0.1
print(f"Hyperparameters: {batch_size=}, {dropout=}")
````

## 5. Lab Exercise: Structured Table Logger

Combine `sep`, `end`, and f-strings to write a reusable console table printer that outputs formatted training metrics.


````python
def log_training_row(epoch: int, loss: float, val_loss: float, f1: float) -> None:
    """Prints a cleanly aligned training log row using f-strings and column specifiers."""
    print(
        f"| Epoch {epoch:03d} | Train Loss: {loss:7.4f} | Val Loss: {val_loss:7.4f} | F1: {f1:6.2%} |"
    )


# Header
print("+" + "-" * 62 + "+")
print(f"| {'EPOCH':<9} | {'TRAIN LOSS':<18} | {'VAL LOSS':<16} | {'F1-SCORE':<8} |")
print("+" + "-" * 62 + "+")

# Log samples
log_training_row(1, 0.6931, 0.6812, 0.5420)
log_training_row(2, 0.5214, 0.4988, 0.7250)
log_training_row(3, 0.3842, 0.3610, 0.8510)
print("+" + "-" * 62 + "+")
````
