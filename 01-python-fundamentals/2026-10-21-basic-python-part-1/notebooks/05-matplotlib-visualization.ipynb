# Visualization with Matplotlib

Matplotlib turns numerical results into figures. A useful plot has a title, labeled axes, a readable scale, and a legend when more than one series is shown.

## Learning goals

- Create a line plot with `matplotlib.pyplot`.
- Label axes, add a title, grid, markers, and a legend.
- Plot the sigmoid and hyperbolic tangent functions.
- Visualize the convergence of a geometric approximation of pi.


## 1. A simple line plot

The `pyplot` interface keeps a current figure and axes. For anything beyond quick exploration, the object-oriented interface (`figure, axis = plt.subplots()`) is clearer because it names the objects you are modifying.


```python
import matplotlib.pyplot as plt

x_values = [0, 1, 2, 3]
y_values = [1, 3, 4, 2]

figure, axis = plt.subplots(figsize=(6, 4))
axis.plot(x_values, y_values, marker="o", label="sample series")
axis.set_title("A simple line plot")
axis.set_xlabel("x")
axis.set_ylabel("y")
axis.set_xticks(x_values)
axis.grid(True, alpha=0.3)
axis.legend()
plt.show()
```

## 2. Activation functions

The sigmoid and hyperbolic tangent are S-shaped activation functions:

$$
\sigma(x) = \frac{1}{1 + e^{-x}},
\qquad
\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}.
$$

The sigmoid maps $\mathbb{R}$ into $(0, 1)$; `tanh` maps $\mathbb{R}$ into $(-1, 1)$. Their derivatives are $\sigma'(x) = \sigma(x)(1 - \sigma(x))$ and $\tanh'(x) = 1 - \tanh^2(x)$.


```python
import numpy as np

x = np.linspace(-6.0, 6.0, 400)
sigmoid = 1.0 / (1.0 + np.exp(-x))
tanh = np.tanh(x)

figure, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot(x, sigmoid, label="sigmoid")
axes[0].plot(x, sigmoid * (1 - sigmoid), label="derivative")
axes[0].set_title("Sigmoid and its derivative")
axes[0].set_xlabel("x")
axes[0].set_ylabel("value")
axes[0].grid(True, alpha=0.3)
axes[0].legend()

axes[1].plot(x, tanh, label="tanh")
axes[1].plot(x, 1 - tanh**2, label="derivative")
axes[1].set_title("Tanh and its derivative")
axes[1].set_xlabel("x")
axes[1].set_ylabel("value")
axes[1].grid(True, alpha=0.3)
axes[1].legend()

figure.tight_layout()
plt.show()
```

## 3. Convergence of an inscribed-polygon estimate of pi

A regular $m$-gon inscribed in a unit circle has side length $2\sin(\pi/m)$ and perimeter $2m\sin(\pi/m)$. Therefore

$$
\pi \approx m \sin\left(\frac{\pi}{m}\right).
$$

The approximation improves as $m$ grows. Plotting the error makes the convergence visible.


```python
segment_counts = np.arange(3, 201)
pi_estimates = segment_counts * np.sin(np.pi / segment_counts)
errors = np.abs(pi_estimates - np.pi)

figure, axis = plt.subplots(figsize=(7, 4))
axis.plot(segment_counts, errors, label="absolute error")
axis.set_title("Inscribed-polygon estimate of pi")
axis.set_xlabel("number of sides")
axis.set_ylabel("absolute error")
axis.set_yscale("log")
axis.grid(True, alpha=0.3, which="both")
axis.legend()
plt.show()

print(f"m = 200: {pi_estimates[-1]:.12f}")
print(f"pi:       {np.pi:.12f}")
```
