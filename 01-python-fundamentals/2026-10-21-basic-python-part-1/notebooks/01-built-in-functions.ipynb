# Built-in Functions, Input, and Type Conversion

Python ships with a small set of built-in functions that cover the most common operations: displaying values, inspecting types, reading user input, converting between types, and computing simple aggregates.

## Learning goals

- Use `print`, `type`, `input`, `int`, and `float` correctly.
- Explain why `input()` always returns a string.
- Convert between strings, integers, and floating-point numbers.
- Apply built-ins to temperature, angle, and geometry problems.
- Avoid comparing floating-point values with exact equality.


## 1. Values and types

Every Python object has a type. The type determines which operations are valid and how the value is represented in memory. The four scalar types used most often in this module are `int`, `float`, `str`, and `bool`.


```python
number_of_days = 10
distance_km = 20.5
greeting = "Hello, learner"
is_enrolled = True

print(number_of_days)
print(distance_km)
print(greeting)
print(is_enrolled)
```

`type(value)` returns the class of a value. Use it while exploring data in a notebook; in production code, prefer `isinstance(value, SomeType)` when a check must accept subclasses.


```python
print(type(number_of_days))
print(type(distance_km))
print(type(greeting))
print(type(is_enrolled))
```

## 2. Reading input

`input(prompt)` writes a prompt and returns the user's response as a **string**, even when the user types digits. Convert the string explicitly when you need a number.

The cell below uses a literal so that the notebook can run without blocking. Replace `value_text = "42"` with `value_text = input("Value: ")` to try the interactive version.


```python
# Interactive alternative:
# value_text = input("Value: ")
value_text = "42"

print(value_text)
print(type(value_text))
```

## 3. Explicit type conversion

`int()` truncates a floating-point number toward zero and parses integer strings. `float()` parses decimal strings and converts integers to floating point. Invalid input raises `ValueError`; handle it when the input comes from a user.


```python
value_int = int(value_text)
value_float = float(value_text)

print(value_int, type(value_int))
print(value_float, type(value_float))

print(int(3.9))
print(float(7))
```

## 4. Angle conversion

The `math` module works in radians. Convert degrees to radians with `math.radians(degrees)` instead of hard-coding an approximation of pi.


```python
import math

degrees = 45.0
radians = math.radians(degrees)

print(f"{degrees} degrees = {radians:.6f} radians")
print(f"sin({degrees} degrees) = {math.sin(radians):.6f}")
```

## 5. Temperature conversion

The conversion between Celsius and Fahrenheit is linear:

$$
F = \frac{9}{5}C + 32,
\qquad
C = \frac{5}{9}(F - 32).
$$


```python
celsius = 37.0
fahrenheit = (9 / 5) * celsius + 32
print(f"{celsius:.1f} degrees Celsius = {fahrenheit:.1f} degrees Fahrenheit")

fahrenheit = 98.6
celsius = (fahrenheit - 32) * 5 / 9
print(f"{fahrenheit:.1f} degrees Fahrenheit = {celsius:.1f} degrees Celsius")
```

## 6. A geometry check with floating-point care

A triangle is right-angled when the square of its longest side equals the sum of the squares of the other two sides. Comparing floats with `==` is fragile because decimal values are stored in binary. Use `math.isclose` with a tolerance instead.


```python
side_a = 3.0
side_b = 4.0
side_c = 5.0

is_right_triangle = math.isclose(side_a**2 + side_b**2, side_c**2)
print(f"Sides {side_a}, {side_b}, {side_c}: right triangle = {is_right_triangle}")

print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
```

## 7. Other useful built-ins

These functions appear constantly in data-processing code. They are ordinary Python functions, not methods of a class.


```python
values = [-2.5, 4.0, 1.5]

print(abs(-7))
print(round(3.14159, 2))
print(min(values))
print(max(values))
print(sum(values))
print(len("Python"))
```
