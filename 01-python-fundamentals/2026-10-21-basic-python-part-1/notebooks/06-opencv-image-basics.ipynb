# Image Basics with OpenCV

An image is a numerical array. OpenCV stores color images as three-channel arrays in blue-green-red (BGR) order, while Matplotlib expects red-green-blue (RGB). Converting the channel order before display is the most common fix for images that look wrong.

## Learning goals

- Load an image with `cv2.imread`.
- Inspect shape, data type, and pixel values.
- Convert BGR to RGB for Matplotlib.
- Convert to grayscale, resize, and crop with array slicing.
- Understand the limitations of `cv2.imshow` in headless environments.


## 1. Load an image

The sample image is stored at `../assets/tree.jpg` relative to this notebook. The second argument is an interpolation/read flag; `cv2.IMREAD_COLOR` is the explicit, readable choice. `cv2.imread` returns `None` when the file cannot be read, so check the result before using it.


```python
from pathlib import Path

import cv2

image_path = Path("..") / "assets" / "tree.jpg"
image_bgr = cv2.imread(str(image_path), cv2.IMREAD_COLOR)

if image_bgr is None:
    raise FileNotFoundError(f"Could not read image: {image_path.resolve()}")

print("shape:", image_bgr.shape)
print("dtype:", image_bgr.dtype)
print("minimum pixel:", image_bgr.min())
print("maximum pixel:", image_bgr.max())
```

The shape is `(height, width, channels)`. The data type is `uint8`, so pixel values lie in $[0, 255]$.


## 2. Display the image

Convert from BGR to RGB with `cv2.cvtColor` before passing the array to Matplotlib. Without the conversion, red and blue regions are swapped.


```python
import matplotlib.pyplot as plt

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

figure, axis = plt.subplots(figsize=(8, 6))
axis.imshow(image_rgb)
axis.set_title("Sample image loaded with OpenCV")
axis.axis("off")
plt.show()
```

## 3. Grayscale conversion

Grayscale reduces the image to a single channel. The weighted conversion reflects human perception: green contributes more than red, and red contributes more than blue.


```python
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)

figure, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(image_rgb)
axes[0].set_title("RGB")
axes[0].axis("off")
axes[1].imshow(image_gray, cmap="gray")
axes[1].set_title("Grayscale")
axes[1].axis("off")
figure.tight_layout()
plt.show()
```

## 4. Resize and crop

OpenCV expects a `(width, height)` tuple in `cv2.resize`, which is the reverse of the array shape order. Array slicing uses `rows` first, then `columns`, so a crop is `image[top:bottom, left:right]`.


```python
height, width = image_bgr.shape[:2]
resized = cv2.resize(
    image_bgr,
    (width // 2, height // 2),
    interpolation=cv2.INTER_AREA,
)
cropped = image_bgr[: height // 2, : width // 2]

print("original:", image_bgr.shape)
print("resized:", resized.shape)
print("cropped:", cropped.shape)

figure, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
axes[0].set_title("Resized to half size")
axes[0].axis("off")
axes[1].imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
axes[1].set_title("Top-left crop")
axes[1].axis("off")
figure.tight_layout()
plt.show()
```
