# The `math` and `random` Modules

The standard library provides mathematical functions and pseudo-random number generation without external dependencies. Both modules are deterministic when their inputs are fixed; `random` becomes reproducible when you set a seed.

## Learning goals

- Use common `math` functions and constants.
- Generate reproducible random values with `random.seed`.
- Draw floating-point and integer samples from the correct intervals.
- Implement a correct sign function.
- Distinguish true division, floor division, and modulo.


## 1. Mathematical functions

Most functions accept and return floats. `math.fabs` always returns a float, while the built-in `abs` preserves the input type for integers.


```python
import math

print("fabs(-2) =", math.fabs(-2))
print("exp(2) =", math.exp(2))
print("log(4) =", math.log(4))
print("log(e) =", math.log(math.e))
print("sqrt(4) =", math.sqrt(4))
print("sin(2) =", math.sin(2))
print("cos(2) =", math.cos(2))
print("pi =", math.pi)
print("e =", math.e)
```

`math.log(x)` is the natural logarithm. Use `math.log(x, base)` for another base. `math.exp(x)` computes $e^x$ and grows extremely quickly, which matters for the numerical-stability notebook.


```python
print("log10(1000) =", math.log(1000, 10))
print("exp(0) =", math.exp(0))
print("sqrt(2) =", math.sqrt(2))
```

## 2. Reproducible randomness

Random numbers are useful for simulations, sampling, and initialization. A seed makes a sequence repeatable, which is essential for debugging and reproducible experiments.

- `random.random()` returns a float in $[0, 1)$.
- `random.randint(a, b)` returns an integer in $[a, b]$, including both endpoints.
- `random.uniform(a, b)` returns a float in $[a, b]$.
- `random.choice(sequence)` selects one element uniformly.


```python
import random

random.seed(42)

print(random.random())
print(random.random())
print(random.randint(1, 9))
print(random.randint(1, 9))
print(random.uniform(-1.0, 1.0))
print(random.choice(["red", "green", "blue"]))
```

Running the cell again with the same seed produces the same sequence. Remove or change the seed when you need independent samples.


## 3. A correct sign function

A common shortcut is `abs(number) / number`, but it fails for zero and obscures the intent. Prefer explicit branches, or use `math.copysign` when only the sign of a nonzero value is needed.


```python
def sign(number: float) -> int:
    """Return -1, 0, or 1 according to the sign of number."""
    if number > 0:
        return 1
    if number < 0:
        return -1
    return 0


for value in (-12.0, 0.0, 7.5):
    print(value, "->", sign(value))
```

## 4. Division and modulo

Python has two division operators:

- `/` performs true division and always returns a float.
- `//` performs floor division and rounds toward negative infinity.

The modulo operator `%` returns the remainder and follows the sign of the divisor. `divmod(a, b)` returns both the quotient and the remainder in one call.


```python
numerator = 9
denominator = 4

print("true division:", numerator / denominator)
print("floor division:", numerator // denominator)
print("remainder:", numerator % denominator)
print("divmod:", divmod(numerator, denominator))

print("-9 // 4 =", -9 // 4)
print("-9 % 4 =", -9 % 4)
```
