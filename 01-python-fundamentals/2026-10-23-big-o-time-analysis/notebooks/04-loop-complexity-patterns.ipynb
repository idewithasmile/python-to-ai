# Loop Complexity Patterns and Step Counting

Loops are the primary source of computational complexity in imperative programs. In this notebook, we analyze the iteration counts and Big-O complexity of standard loop structures.

## Learning Outcomes

- Analyze single loops with unit step ($O(n)$) and geometric step ($O(\log n)$).
- Compute exact iteration counts for base-2 and base-3 multiplicative loops.
- Analyze nested rectangular loops ($O(n \times m)$) and nested triangular loops ($O(n^2)$).
- Derive the arithmetic progression sum $\sum_{i=1}^n i = \frac{n(n+1)}{2}$.

## 1. Single Loops: Unit Step vs. Multiplicative Step

In a unit step loop, $i$ advances by a constant $+1$ at each iteration $\implies n$ iterations ($O(n)$).

In a multiplicative loop:
```python
i = 1
while i < n:
    i *= 2
```
Values taken by $i$: $1, 2, 4, 8, \dots, 2^k$.
The loop halts when $2^k \ge n \iff k \ge \log_2 n$.
Thus, the number of body executions is $\lceil \log_2 n \rceil$, yielding $O(\log_2 n)$ complexity.

````python
import math


def step_example_base2(n: int) -> int:
    """Count body executions for i *= 2 loop."""
    i = 1
    count = 0
    while i < n:
        count += 1
        i *= 2
    return count


def step_example_base3(n: int) -> int:
    """Count body executions for i *= 3 loop."""
    i = 1
    count = 0
    while i < n:
        count += 1
        i *= 3
    return count


for test_n in [1, 2, 3, 4, 8, 16, 27, 32, 64, 81, 100]:
    c2 = step_example_base2(test_n)
    c3 = step_example_base3(test_n)
    # Expected base-2 iterations: ceil(log2(test_n)) if test_n > 1 else 0
    exp2 = math.ceil(math.log2(test_n)) if test_n > 1 else 0
    exp3 = math.ceil(math.log(test_n, 3)) if test_n > 1 else 0
    print(
        f"n = {test_n:3d} | Base-2 steps: {c2:2d} (exp {exp2:2d}) | Base-3 steps: {c3:2d} (exp {exp3:2d})"
    )
    assert c2 == exp2
    assert c3 == exp3
````

## 2. Nested Rectangular Loops ($O(n^2)$)

When both outer and inner loops iterate independently $n$ times:
$$\sum_{i=0}^{n-1} \sum_{j=0}^{n-1} 1 = n \times n = n^2 \implies O(n^2)$$

````python
def nested_rectangular(n: int) -> int:
    iterations = 0
    for _ in range(n):
        for _ in range(n):
            iterations += 1
    return iterations


for n in [1, 2, 5, 10, 20]:
    assert nested_rectangular(n) == n * n
print("Nested rectangular loop verified: exactly n^2 iterations.")
````

## 3. Nested Triangular Loops: Arithmetic Series

Consider a nested loop where the inner loop limit depends on outer loop index $i$:

```python
def sum_example(S):
    n = len(S)
    total = 0
    for i in range(n):
        for j in range(1 + i):
            total += S[j]
    return total
```

The inner loop runs $1, 2, 3, \dots, n$ times.
Total inner iterations:
$$\sum_{i=0}^{n-1} (i + 1) = \sum_{k=1}^n k = \frac{n(n + 1)}{2} = \frac{1}{2}n^2 + \frac{1}{2}n$$

As $n \to \infty$, the $\frac{1}{2}n^2$ term dominates, proving this algorithm is $\Theta(n^2)$.

````python
def triangular_loop_iterations(n: int) -> int:
    count = 0
    for i in range(n):
        for _ in range(i + 1):
            count += 1
    return count


for n in [1, 2, 3, 4, 10, 50, 100]:
    actual = triangular_loop_iterations(n)
    expected = (n * (n + 1)) // 2
    print(f"n = {n:3d} | Actual inner iterations: {actual:5d} | n(n+1)/2: {expected:5d}")
    assert actual == expected
````

## 4. Nested Loop with Logarithmic Inner Loop ($O(n \log n)$)

If an outer loop runs $n$ times, and the inner loop multiplies by $2$ until $n$:
$$\sum_{i=1}^n \lceil \log_2 n \rceil = n \cdot \lceil \log_2 n \rceil \implies O(n \log n)$$

````python
import math


def n_logn_loop_iterations(n: int) -> int:
    count = 0
    for _ in range(n):
        j = 1
        while j < n:
            count += 1
            j *= 2
    return count


for n in [2, 4, 8, 16, 32, 64]:
    actual = n_logn_loop_iterations(n)
    expected = n * math.ceil(math.log2(n))
    print(f"n = {n:2d} | Actual: {actual:4d} | Expected n*ceil(log2 n): {expected:4d}")
    assert actual == expected
````
