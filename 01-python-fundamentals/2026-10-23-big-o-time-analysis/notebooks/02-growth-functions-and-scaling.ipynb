# The Seven Fundamental Growth Functions and Scaling

When analyzing algorithms, seven core mathematical functions describe the growth rate of runtime as input size $n$ increases to infinity.

## Learning Outcomes

- Master the seven fundamental growth functions: constant, logarithmic, linear, linearithmic, quadratic, cubic, and exponential.
- Understand key logarithmic properties and why base does not alter asymptotic growth.
- Build numerical scaling tables comparing execution operations across sizes $n = 10^1$ to $n = 10^9$.
- Plot growth trajectories on linear, semilog, and log-log scales using Matplotlib.

## 1. Mathematical Definitions of the Seven Functions

1. **Constant function**: $f(n) = c$ ($O(1)$)
2. **Logarithmic function**: $f(n) = \log_b n$ ($O(\log n)$)
3. **Linear function**: $f(n) = n$ ($O(n)$)
4. **Linearithmic (N-Log-N) function**: $f(n) = n \log_b n$ ($O(n \log n)$)
5. **Quadratic function**: $f(n) = n^2$ ($O(n^2)$)
6. **Cubic function**: $f(n) = n^3$ ($O(n^3)$)
7. **Exponential function**: $f(n) = b^n$ ($O(b^n)$, $b > 1$)

```python
import matplotlib.pyplot as plt
import numpy as np


# Define seven functions for evaluation
def f_constant(n: np.ndarray) -> np.ndarray:
    return np.ones_like(n, dtype=float)


def f_log(n: np.ndarray) -> np.ndarray:
    return np.log2(np.maximum(n, 1.0))


def f_linear(n: np.ndarray) -> np.ndarray:
    return n.astype(float)


def f_nlogn(n: np.ndarray) -> np.ndarray:
    return n * np.log2(np.maximum(n, 1.0))


def f_quadratic(n: np.ndarray) -> np.ndarray:
    return n.astype(float) ** 2


def f_cubic(n: np.ndarray) -> np.ndarray:
    return n.astype(float) ** 3


def f_exponential(n: np.ndarray) -> np.ndarray:
    return 2.0 ** np.minimum(n, 30)  # capped to avoid float overflow


print("Defined functions: constant, log2, linear, nlogn, quadratic, cubic, exponential.")
```

## 2. Essential Properties of Logarithms

Logarithms invert exponentiation ($b^y = x \iff \log_b x = y$). Key identities:

- $\log_b (a \cdot c) = \log_b a + \log_b c$
- $\log_b (a / c) = \log_b a - \log_b c$
- $\log_b (a^k) = k \log_b a$
- **Change of base formula**:
  $$\log_b a = \frac{\log_c a}{\log_c b}$$

Because $\frac{1}{\log_c b}$ is a constant scalar, changing the base of a logarithm only scales the function by a constant factor. Thus, $O(\log_2 n) = O(\log_{10} n) = O(\ln n) = O(\log n)$.

```python
import math

# Verifying log laws programmatically
a, c = 16.0, 4.0
assert math.isclose(math.log2(a * c), math.log2(a) + math.log2(c))
assert math.isclose(math.log2(a / c), math.log2(a) - math.log2(c))
assert math.isclose(math.log2(a**3), 3 * math.log2(a))

# Change of base: log2(100) vs log10(100)/log10(2)
val = 100.0
assert math.isclose(math.log2(val), math.log10(val) / math.log10(2))
print("All logarithm identities verified numerically!")
```

## 3. Numerical Scaling Comparison Across Input Sizes

Let us observe how each function behaves as $n$ scales from $10$ to $10^9$.

```python
import math

input_sizes = [1, 2, 4, 8, 16, 32, 64, 128]
header = f"{'n':>5} | {'c':>5} | {'log2(n)':>8} | {'n':>5} | {'n log2(n)':>10} | {'n^2':>8} | {'n^3':>10} | {'2^n':>12}"
print(header)
print("-" * len(header))

for n in input_sizes:
    c_val = 1
    log_val = math.log2(n) if n > 0 else 0
    lin_val = n
    nlogn_val = n * log_val
    quad_val = n**2
    cube_val = n**3
    exp_val = 2**n if n <= 32 else float("inf")
    exp_str = f"{exp_val:12.0f}" if n <= 32 else "  > 10^38   "
    print(
        f"{n:5d} | {c_val:5d} | {log_val:8.2f} | {lin_val:5d} | {nlogn_val:10.2f} | {quad_val:8d} | {cube_val:10d} | {exp_str}"
    )
```

## 4. Visualizing Growth Trajectories with Matplotlib

We plot the seven growth functions across $n \in [1, 50]$.

```python
n_vals = np.linspace(1, 40, 300)

plt.figure(figsize=(10, 6))
plt.plot(n_vals, f_constant(n_vals), label="O(1) Constant", color="gray", linestyle="--")
plt.plot(n_vals, f_log(n_vals), label="O(log n) Logarithmic", color="blue")
plt.plot(n_vals, f_linear(n_vals), label="O(n) Linear", color="green")
plt.plot(n_vals, f_nlogn(n_vals), label="O(n log n) Linearithmic", color="purple")
plt.plot(n_vals, f_quadratic(n_vals), label="O(n^2) Quadratic", color="orange")
plt.plot(n_vals, f_cubic(n_vals), label="O(n^3) Cubic", color="red")

plt.ylim(0, 1500)
plt.xlim(1, 40)
plt.xlabel("Input Size n", fontsize=12)
plt.ylabel("Operations f(n)", fontsize=12)
plt.title("Asymptotic Growth of Fundamental Complexity Functions", fontsize=14)
plt.legend(loc="upper left", frameon=True)
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 5. Semi-Log and Log-Log Visualizations

On a logarithmic y-axis, polynomial functions curve downward, while exponential functions appear as straight lines. On a log-log plot, polynomial functions $f(n) = n^k$ appear as straight lines with slope equal to their degree $k$.

```python
n_poly = np.logspace(0, 3, 200)  # n from 1 to 1000

plt.figure(figsize=(10, 6))
plt.plot(n_poly, f_log(n_poly), label="log2(n)", color="blue")
plt.plot(n_poly, f_linear(n_poly), label="n (slope=1)", color="green")
plt.plot(n_poly, f_nlogn(n_poly), label="n log2(n)", color="purple")
plt.plot(n_poly, f_quadratic(n_poly), label="n^2 (slope=2)", color="orange")
plt.plot(n_poly, f_cubic(n_poly), label="n^3 (slope=3)", color="red")

plt.xscale("log")
plt.yscale("log")
plt.xlabel("log(n)", fontsize=12)
plt.ylabel("log(f(n))", fontsize=12)
plt.title("Log-Log Scaling: Slopes Reveal Polynomial Degrees", fontsize=14)
plt.legend(loc="upper left")
plt.grid(True, which="both", linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```
