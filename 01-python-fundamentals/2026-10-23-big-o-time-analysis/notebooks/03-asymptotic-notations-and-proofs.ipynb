# Asymptotic Notations and Formal Bounding Proofs

Asymptotic analysis characterizes the limiting behavior of functions as $n \to \infty$. We study Big-O ($O$), Big-Omega ($\Omega$), and Big-Theta ($\Theta$) notations with formal proofs and computational verification.

## Learning Outcomes

- Master the formal definitions of Big-O, Big-Omega, and Big-Theta.
- Prove asymptotic membership by finding concrete constants $c$ and $n_0$.
- Understand properties: dropping constants, leading terms, transitivity, sums, and products.
- Verify asymptotic bounds programmatically with Python assertion tests.

## 1. Formal Definitions of Asymptotic Notations

### Big-O Notation (Asymptotic Upper Bound)
$$f(n) \in O(g(n)) \iff \exists c > 0, n_0 \ge 1 \text{ such that } \forall n \ge n_0, f(n) \le c \cdot g(n)$$

### Big-Omega Notation (Asymptotic Lower Bound)
$$f(n) \in \Omega(g(n)) \iff \exists c > 0, n_0 \ge 1 \text{ such that } \forall n \ge n_0, f(n) \ge c \cdot g(n)$$

### Big-Theta Notation (Asymptotically Tight Bound)
$$f(n) \in \Theta(g(n)) \iff \exists c_1, c_2 > 0, n_0 \ge 1 \text{ such that } \forall n \ge n_0, c_1 g(n) \le f(n) \le c_2 g(n)$$
Equivalently: $f(n) \in \Theta(g(n)) \iff f(n) \in O(g(n)) \text{ and } f(n) \in \Omega(g(n))$.

```python
from collections.abc import Callable


def verify_big_o(
    f: Callable[[int], float], g: Callable[[int], float], c: float, n0: int, max_n: int = 1000
) -> bool:
    """Verify that f(n) <= c * g(n) for all n from n0 to max_n."""
    return all(f(n) <= c * g(n) for n in range(n0, max_n + 1))


def verify_big_omega(
    f: Callable[[int], float], g: Callable[[int], float], c: float, n0: int, max_n: int = 1000
) -> bool:
    """Verify that f(n) >= c * g(n) for all n from n0 to max_n."""
    return all(f(n) >= c * g(n) for n in range(n0, max_n + 1))


def verify_big_theta(
    f: Callable[[int], float],
    g: Callable[[int], float],
    c1: float,
    c2: float,
    n0: int,
    max_n: int = 1000,
) -> bool:
    """Verify that c1 * g(n) <= f(n) <= c2 * g(n) for all n from n0 to max_n."""
    return verify_big_omega(f, g, c1, n0, max_n) and verify_big_o(f, g, c2, n0, max_n)


print("Verification functions loaded.")
```

## 2. Formal Proofs: Big-O Upper Bounds

### Proof 1: $f(n) = 8n + 5 \in O(n)$
For all $n \ge 1$:
$$8n + 5 \le 8n + 5n = 13n$$
Choose $c = 13$ and $n_0 = 1$. Then $8n + 5 \le 13n$ holds for all $n \ge 1$.

### Proof 2: $f(n) = (n + 1)^5 \in O(n^5)$
Expanding by the binomial theorem:
$$(n + 1)^5 = n^5 + 5n^4 + 10n^3 + 10n^2 + 5n + 1$$
Since $n^k \le n^5$ for all $n \ge 1$ and $k \le 5$:
$$(n + 1)^5 \le (1 + 5 + 10 + 10 + 5 + 1)n^5 = 32 n^5$$
Choose $c = 32$ and $n_0 = 1$.

```python
import math

# Verify Proof 1 and 2
assert verify_big_o(lambda n: 8 * n + 5, lambda n: n, c=13, n0=1)
assert verify_big_o(lambda n: (n + 1) ** 5, lambda n: n**5, c=32, n0=1)

# Proof 3: 2^(n+1) is O(2^n)
# 2^(n+1) = 2 * 2^n -> c = 2, n0 = 1
assert verify_big_o(lambda n: 2 ** (n + 1), lambda n: 2**n, c=2, n0=1, max_n=30)

# Proof 4: 5n^2 + 3n log2(n) + 2n + 5 is O(n^2)
# 5n^2 + 3n log2(n) + 2n + 5 <= (5 + 3 + 2 + 5)n^2 = 15 n^2 for n >= 1
assert verify_big_o(
    lambda n: 5 * (n**2) + 3 * n * math.log2(n) + 2 * n + 5 if n > 0 else 0,
    lambda n: n**2,
    c=15,
    n0=1,
)
print("Proofs 1-4 verified successfully!")
```

## 3. Formal Proofs: Big-Omega and Big-Theta

### Proof 5: $f(n) = 3n\log_2 n - 2n \in \Omega(n\log_2 n)$
Notice:
$$3n\log_2 n - 2n = n\log_2 n + 2n(\log_2 n - 1)$$
For $n \ge 2$, $\log_2 n \ge 1$, so $2n(\log_2 n - 1) \ge 0$.
Therefore:
$$3n\log_2 n - 2n \ge n\log_2 n = 1 \cdot n\log_2 n$$
Choose $c = 1, n_0 = 2$.

### Proof 6: $f(n) = 3n\log_2 n + 4n + 5\log_2 n \in \Theta(n\log_2 n)$
- Lower bound: For $n \ge 2$, all terms are positive, so $f(n) \ge 3n\log_2 n \to c_1 = 3, n_0 = 2$.
- Upper bound: For $n \ge 2$, $4n \le 4n\log_2 n$ and $5\log_2 n \le 5n\log_2 n$.
  $$f(n) \le (3 + 4 + 5)n\log_2 n = 12n\log_2 n \to c_2 = 12, n_0 = 2$$.

```python
# Verify Proof 5: Big-Omega
assert verify_big_omega(
    lambda n: 3 * n * math.log2(n) - 2 * n,
    lambda n: n * math.log2(n),
    c=1.0,
    n0=2,
)

# Verify Proof 6: Big-Theta
assert verify_big_theta(
    lambda n: 3 * n * math.log2(n) + 4 * n + 5 * math.log2(n),
    lambda n: n * math.log2(n),
    c1=3.0,
    c2=12.0,
    n0=2,
)
print("Proofs 5 and 6 (Big-Omega and Big-Theta) verified!")
```

## 4. Asymptotic Ranking and Growth Order

Ranking functions by asymptotic dominance:

$$O(1) \subset O(\log n) \subset O(n) \subset O(n\log n) \subset O(n^2) \subset O(n^3) \subset O(2^n)$$

Let us test tricky function pairs:
- $2^{\log_2 n} = n \implies O(n)$
- $4^{\log_2 n} = (2^2)^{\log_2 n} = (2^{\log_2 n})^2 = n^2 \implies O(n^2)$
- $n^{1/\log_2 n} = 2^{(\log_2 n)(1/\log_2 n)} = 2^1 = 2 \implies O(1)$!

```python
for n in [8, 16, 32, 64, 128]:
    # Test 2^(log2 n)
    v1 = 2 ** math.log2(n)
    assert math.isclose(v1, n)

    # Test 4^(log2 n)
    v2 = 4 ** math.log2(n)
    assert math.isclose(v2, n**2)

    # Test n^(1/log2 n)
    v3 = n ** (1.0 / math.log2(n))
    assert math.isclose(v3, 2.0)

print("All algebraic asymptotic equivalences confirmed!")
```
