**Bubble Sort** operates by scanning adjacent pairs of elements across the sequence and swapping them if they violate monotonic order. Through repeated passes, the largest unsorted value "bubbles up" to the end of the array.

## Learning Outcomes

- Implement standard and optimized adaptive Bubble Sort.
- Understand the early termination condition (`swapped = False`).
- Characterize the mathematical connection between swaps and permutation inversions.
- Contrast Bubble Sort with Selection Sort in terms of cache behavior and stability.


## 1. Unoptimized vs Adaptive Bubble Sort

Standard Bubble Sort unconditionally executes $n-1$ passes. By introducing a boolean flag `swapped`, the algorithm terminates immediately if a full pass completes without any swaps, reducing best-case runtime to $O(n)$.


```python
def bubble_sort_unoptimized(array: list[int]) -> tuple[list[int], int]:
    """Standard Bubble Sort without early termination."""
    a = array.copy()
    n = len(a)
    comparisons = 0
    for step in range(n):
        for i in range(0, n - step - 1):
            comparisons += 1
            if a[i] > a[i + 1]:
                a[i], a[i + 1] = a[i + 1], a[i]
    return a, comparisons


def bubble_sort_optimized(array: list[int]) -> tuple[list[int], int]:
    """Adaptive Bubble Sort with early stopping flag."""
    a = array.copy()
    n = len(a)
    comparisons = 0
    for step in range(n):
        swapped = False
        for i in range(0, n - step - 1):
            comparisons += 1
            if a[i] > a[i + 1]:
                a[i], a[i + 1] = a[i + 1], a[i]
                swapped = True
        if not swapped:
            break
    return a, comparisons


# Lecture test case
data = [7, 2, 5, 4, 17]
sorted_arr, comps = bubble_sort_optimized(data)
print("Sorted:", sorted_arr)
print("Comparisons on [7, 2, 5, 4, 17]:", comps)
assert sorted_arr == [2, 4, 5, 7, 17]
```

## 2. Inversions and Swap Count

An **inversion** in a sequence $A$ is a pair of indices $(i, j)$ such that $i < j$ and $A[i] > A[j]$.

Each adjacent swap in Bubble Sort resolves exactly one inversion. Hence, the total number of swaps executed by Bubble Sort is **strictly equal** to the inversion count of the input array.


```python
def count_inversions(arr: list[int]) -> int:
    """Count total number of inversions by definition."""
    inversions = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] > arr[j]:
                inversions += 1
    return inversions


def count_bubble_swaps(arr: list[int]) -> int:
    """Count swaps in Bubble Sort."""
    a = arr.copy()
    n = len(a)
    swaps = 0
    for step in range(n):
        for i in range(0, n - step - 1):
            if a[i] > a[i + 1]:
                a[i], a[i + 1] = a[i + 1], a[i]
                swaps += 1
    return swaps


test_cases = [[7, 2, 5, 4, 17], [5, 4, 3, 2, 1], [1, 2, 3, 4, 5], [3, 1, 2]]

for tc in test_cases:
    inv = count_inversions(tc)
    swp = count_bubble_swaps(tc)
    print(f"Array {tc}: Inversions = {inv}, Bubble Swaps = {swp}")
    assert inv == swp
```

## 3. Complexity Comparison

| Metric | Selection Sort | Bubble Sort (Adaptive) |
| :--- | :--- | :--- |
| **Best Case Time** | $O(n^2)$ | $O(n)$ |
| **Average Case Time** | $O(n^2)$ | $O(n^2)$ |
| **Worst Case Time** | $O(n^2)$ | $O(n^2)$ |
| **Worst Case Swaps** | $O(n)$ | $O(n^2)$ |
| **Stability** | Unstable | **Stable** (strictly adjacent swaps) |
| **Memory Writes** | Minimal ($n - 1$) | High (proportional to inversions) |


```python
# Benchmark best-case comparison
sorted_array = list(range(100))
_, unopt_comps = bubble_sort_unoptimized(sorted_array)
_, opt_comps = bubble_sort_optimized(sorted_array)

print("Already sorted array (n=100):")
print(f"Unoptimized comparisons: {unopt_comps} (n(n-1)/2)")
print(f"Optimized comparisons:   {opt_comps} (n-1)")
assert opt_comps == 99
```
