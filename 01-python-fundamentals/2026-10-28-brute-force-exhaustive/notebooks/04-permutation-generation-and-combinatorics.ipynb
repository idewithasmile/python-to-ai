**Exhaustive Search** operates over discrete solution spaces. When candidates represent orderings, the search space consists of all permutations of $n$ elements ($n!$). When candidates represent selections, the search space is the power set ($2^n$).

## 1. Iterative Permutation Generation

The source course material introduced an iterative insertion strategy:
1. Start with $L = [[x_0]]$.
2. For each subsequent element $x_i$, insert $x_i$ into all possible positions $j \in [0 \dots i]$ of each existing permutation in $L$.


```python
from typing import Any


def permute_iterative(nums: list[Any]) -> list[list[Any]]:
    """Iteratively generate all permutations by positional insertion."""
    result: list[list[Any]] = []
    n = len(nums)
    i = 0
    while i < n:
        if not result:
            result.append([nums[i]])
        else:
            new_perms: list[list[Any]] = []
            for perm in result:
                for j in range(i + 1):
                    copy = perm.copy()
                    copy.insert(j, nums[i])
                    new_perms.append(copy)
            result = new_perms
        i += 1
    return result


sample = [1, 2, 3]
perms = permute_iterative(sample)
print(f"Permutations of {sample} (total = {len(perms)}):")
for p in perms:
    print(" ", p)

assert len(perms) == 6
assert [1, 2, 3] in perms
assert [3, 2, 1] in perms
```

## 2. Recursive Backtracking Permutation Generator

A classic, in-place depth-first search approach generates permutations by swapping elements:
- Fix index `start`.
- Swap each candidate $i \in [\text{start} \dots n-1]$ into `start`.
- Recurse on $\text{start} + 1$.
- Backtrack by undoing the swap.


```python
def permute_backtrack(nums: list[Any]) -> list[list[Any]]:
    """Generate permutations using recursive backtracking."""
    result: list[list[Any]] = []
    arr = nums.copy()

    def backtrack(start: int):
        if start == len(arr):
            result.append(arr.copy())
            return
        for i in range(start, len(arr)):
            arr[start], arr[i] = arr[i], arr[start]
            backtrack(start + 1)
            arr[start], arr[i] = arr[i], arr[start]

    backtrack(0)
    return result


res_backtrack = permute_backtrack([1, 2, 3])
assert len(res_backtrack) == 6
print("Backtracking generator produced 6 valid permutations.")
```

## 3. Power Set Generation ($2^n$ Subsets)

Many exhaustive search problems (such as the Knapsack problem or Feature Selection) explore subsets rather than permutations. The power set of a set with $n$ elements contains exactly $2^n$ subsets.


```python
def generate_power_set(items: list[Any]) -> list[list[Any]]:
    """Generate all 2^n subsets using binary mask representation."""
    n = len(items)
    subsets = []
    total_subsets = 1 << n  # 2^n

    for mask in range(total_subsets):
        subset = [items[i] for i in range(n) if (mask & (1 << i)) != 0]
        subsets.append(subset)

    return subsets


items = ["a", "b", "c"]
pset = generate_power_set(items)
print(f"Power set of {items} has {len(pset)} subsets:")
for s in pset:
    print(" ", s)
assert len(pset) == 8
```

## 4. Combinatorial Growth Comparison

The following table contrasts polynomial ($n^2$), exponential ($2^n$), and factorial ($n!$) growth rates:

| $n$ | $n^2$ | $2^n$ | $n!$ |
| :--- | :--- | :--- | :--- |
| 4 | 16 | 16 | 24 |
| 8 | 64 | 256 | 40,320 |
| 10 | 100 | 1,024 | 3,628,800 |
| 12 | 144 | 4,096 | 479,001,600 |
| 15 | 225 | 32,768 | 1,307,674,368,000 |

Factorial spaces quickly exceed the computational throughput of modern supercomputers.
