Python provides foundational functional primitives that operate directly on sequence protocols without requiring imperative loop state. Leveraging built-ins such as `len()`, `min()`, `max()`, `sum()`, `reversed()`, `enumerate()`, and `zip()` produces idiomatic, readable, and highly optimized code.

### Learning Objectives
1. Contrast built-in reductions (`sum`, `min`, `max`) with manual accumulator patterns.
2. Leverage `reversed()` for $O(1)$-memory lazy sequence reversal.
3. Pair values with coordinate indices cleanly using `enumerate()`.
4. Traverse multiple sequences in lockstep with `zip()`.
5. Implement a continuous linear coordinate generator (`linear_space`).

```
# Sequence Reductions: sum(), min(), max(), len()
samples: list[int] = [6, 5, 7, 1, 9, 2]

# len() is O(1) in CPython
count: int = len(samples)
min_val: int = min(samples)
max_val: int = max(samples)
total: int = sum(samples)

print(f"Count: {count}, Min: {min_val}, Max: {max_val}, Sum: {total}")


# Verification against manual accumulator
def manual_sum(data: list[int]) -> int:
    acc = 0
    for val in data:
        acc += val
    return acc


assert manual_sum(samples) == total == 30
assert min_val == 1 and max_val == 9
print("Reduction assertions passed successfully.")
```

## Sequence Iterators: `reversed`, `enumerate`, and `zip`

1. `reversed(seq)`: Returns a reverse iterator. Unlike `seq[::-1]`, it allocates zero auxiliary memory for elements ($O(1)$ space).
2. `enumerate(iterable, start=0)`: Yields pairs of `(index, item)`, avoiding fragile manual counter variables `i = 0; i += 1`.
3. `zip(*iterables)`: Yields tuples aggregating elements from each of the iterables in lockstep until the shortest iterable is exhausted.

```
# Iteration protocols demonstration
features = ["height", "weight", "age", "blood_pressure"]
coefficients = [0.45, -0.32, 1.20, 0.85]

# enumerate: index-value binding
enumerated_features: list[tuple[int, str]] = list(enumerate(features))
print(f"Enumerated: {enumerated_features}")
assert enumerated_features[0] == (0, "height")

# reversed iterator
rev_features = list(reversed(features))
assert rev_features == ["blood_pressure", "age", "weight", "height"]

# zip: parallel association
model_terms = list(zip(features, coefficients, strict=True))
print(f"Model terms: {model_terms}")
assert model_terms[1] == ("weight", -0.32)
print("Iterator assertions passed successfully.")
```

## Custom Reductions and Numerical Grid Generator (`linear_space`)

In scientific computing, generating linearly spaced coordinate arrays is a foundational requirement for sampling continuous functions $f(x)$ over an interval $[a, b]$.

$$\Delta x = \frac{b - a}{\text{num} - 1}, \quad x_i = a + i \cdot \Delta x \quad \text{for } i = 0, 1, \dots, \text{num} - 1$$

```
# Conditional accumulators and linear space generator
def sum_even_values(data: list[int]) -> int:
    return sum(x for x in data if x % 2 == 0)


def sum_even_indices(data: list[int]) -> int:
    return sum(val for idx, val in enumerate(data) if idx % 2 == 0)


def linear_space(start: float, stop: float, num: int = 10) -> list[float]:
    if num <= 1:
        raise ValueError("num must be strictly greater than 1")
    step = (stop - start) / (num - 1)
    return [start + i * step for i in range(num)]


# Tests
test_data = [6, 5, 7, 1, 9, 2]
assert sum_even_values(test_data) == 6 + 2 == 8
assert sum_even_indices(test_data) == test_data[0] + test_data[2] + test_data[4] == 6 + 7 + 9 == 22

grid = linear_space(-5.0, 5.0, num=11)
assert len(grid) == 11
assert abs(grid[0] - (-5.0)) < 1e-9
assert abs(grid[-1] - 5.0) < 1e-9
assert abs(grid[5] - 0.0) < 1e-9
print("Linear space and conditional accumulators validated.")
```

## Practical Lab Exercise: Dot Product and Cumulative Moving Average

Implement:
1. `dot_product(vec_a: list[float], vec_b: list[float]) -> float`: Computes the algebraic Euclidean dot product $\mathbf{a} \cdot \mathbf{b} = \sum_{i=0}^{n-1} a_i b_i$ using `zip()`.
2. `cumulative_moving_average(series: list[float]) -> list[float]`: Computes running averages $M_k = \frac{1}{k+1} \sum_{i=0}^k x_i$ using `enumerate()`.

```
def dot_product(vec_a: list[float], vec_b: list[float]) -> float:
    if len(vec_a) != len(vec_b):
        raise ValueError("Vectors must have matching dimensional length")
    return sum(x * y for x, y in zip(vec_a, vec_b, strict=True))


def cumulative_moving_average(series: list[float]) -> list[float]:
    running_total = 0.0
    cma = []
    for idx, value in enumerate(series):
        running_total += value
        cma.append(running_total / (idx + 1))
    return cma


# Verify
u = [1.0, 2.0, 3.0]
v = [4.0, 5.0, 6.0]
assert dot_product(u, v) == 1.0 * 4.0 + 2.0 * 5.0 + 3.0 * 6.0 == 32.0

stream = [10.0, 20.0, 30.0, 40.0]
averages = cumulative_moving_average(stream)
assert averages == [10.0, 15.0, 20.0, 25.0]
print("Lab exercise assertions passed successfully!")
```
