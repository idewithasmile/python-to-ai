# For Loops

A `for` loop visits the elements of an iterable and runs a block of code once per element. It is the main tool for traversing ranges, strings, lists, tuples, dictionaries, and files.

## Learning goals

- Use `range` with one, two, and three arguments.
- Traverse strings, lists, tuples, and dictionaries.
- Predict the state of a loop with a trace table.
- Control a loop with `break`, `continue`, and `else`.
- Build factorial, Euler's number, Newton's method, and a coin-tossing simulation.

## 1. The smallest useful loop

`range(5)` produces the integers `0, 1, 2, 3, 4`. The value `5` is not included. This half-open convention is used throughout Python.

````python
for number in range(5):
    print(number)
````

The general form is `range(start, stop, step)`. `start` is inclusive, `stop` is exclusive, and `step` may be negative.

````python
for number in range(2, 6):
    print(number)
````

````python
for number in range(10, 0, -2):
    print(number)
````

A `range` object is lazy and immutable. Membership tests and indexing are constant-time operations, and the object does not store every integer.

````python
numbers = range(0, 100, 5)
print(7 in numbers)
print(numbers[3])
print(len(numbers))
````

## 2. Iterating over collections

A string yields characters, a list yields its elements, and a dictionary yields its keys. Use `items()` to visit keys and values together.

````python
word = "Python"
for character in word:
    print(character.upper())
````

````python
values = [10, 20, 30]
for value in values:
    print(value * 2)
````

````python
config = {"learning_rate": 0.1, "epochs": 20}
for key, value in config.items():
    print(f"{key}: {value}")
````

## 3. Trace tables

Before running a loop, write down the values of the variables that change. The table below describes the following program:

```python
total = 0
for number in range(1, 5):
    total += number
```

| Iteration | `number` | `total` before | `total` after |
| :---: | :---: | :---: | :---: |
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 4 | 6 | 10 |

The final value is `10`. If your trace disagrees with the program, one of the two is wrong; the trace usually reveals which.

````python
total = 0
for number in range(1, 5):
    total += number

print(total)
````

## 4. `break` and `continue`

`break` exits the nearest enclosing loop. `continue` skips the rest of the current iteration. Both statements act on loops, not on `if` blocks.

````python
for number in range(10):
    if number == 5:
        break
    print(number)
````

````python
for number in range(10):
    if number == 5:
        continue
    print(number)
````

The optional `else` block runs only when the loop finishes without a `break`. It is useful for search loops: if no match is found, the `else` block handles the failure.

````python
def contains_negative(values):
    for value in values:
        if value < 0:
            print("Found a negative value")
            break
    else:
        print("All values are non-negative")


contains_negative([3, 1, 4])
contains_negative([3, -1, 4])
````

## 5. Nested loops

The inner loop runs completely for every iteration of the outer loop. The total number of body executions is the product of the iteration counts.

````python
for row in range(1, 6):
    for column in range(1, 6):
        print(f"{row * column:4d}", end="")
    print()
````

A `break` inside a nested loop exits only the inner loop. If you need to leave both loops, return from a function instead of adding flags.

````python
def find_value(matrix, target):
    for row in matrix:
        for value in row:
            if value == target:
                return True
    return False


matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(find_value(matrix, 5))
print(find_value(matrix, 10))
````

## 6. Factorial

The factorial is defined as

$$
n! = \prod_{k=1}^{n} k,
$$

with $0! = 1$. The loop starts at `2` because multiplying by `1` changes nothing.

````python
def factorial(n):
    if n < 0:
        raise ValueError("factorial is undefined for negative integers")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


print(factorial(0))
print(factorial(5))
assert factorial(5) == 120
````

Factorials grow very quickly. Python integers are arbitrary precision, so the integer result is exact, but converting a large factorial to `float` can raise `OverflowError`.

## 7. Euler's number

Euler's number is the infinite series

$$
e = \sum_{k=0}^{\infty} \frac{1}{k!}.
$$

Truncating the series gives a practical approximation. The error decreases rapidly because factorials grow so fast.

````python
import math


def estimate_e(terms):
    total = 0.0
    for k in range(terms):
        total += 1.0 / factorial(k)
    return total


for terms in (5, 10, 15, 20):
    estimate = estimate_e(terms)
    print(f"terms={terms:>2} estimate={estimate:.15f} error={abs(estimate - math.e):.3e}")
````

## 8. Newton's method for square roots

To compute $\sqrt{N}$, solve $x^2 - N = 0$. Newton's update simplifies to

$$
x_{n+1} = \frac{1}{2}\left(x_n + \frac{N}{x_n}\right).
$$

Starting at $x_0 = N/2$ is safe for $N > 0$. The method converges quadratically near the root, so only a few iterations are usually needed.

````python
def sqrt_newton(number, tolerance=1e-12, max_iterations=100):
    if number < 0:
        raise ValueError("number must be non-negative")
    if number == 0:
        return 0.0

    estimate = number / 2.0
    for _ in range(max_iterations):
        next_estimate = (estimate + number / estimate) / 2.0
        print(f"{estimate:.15f}")
        if abs(next_estimate - estimate) < tolerance:
            return next_estimate
        estimate = next_estimate

    raise RuntimeError("Newton's method did not converge")


root = sqrt_newton(9)
print(f"root={root:.15f}")
assert abs(root - 3.0) < 1e-12
````

## 9. Simulating coin tossing

A fair coin lands on heads or tails with equal probability. Draw a uniform random number in $[0, 1)$ and count values below `0.5` as tails. Seeding the generator makes the experiment reproducible.

````python
import random


def simulate_coin_tosses(tosses, seed=0):
    random.seed(seed)
    heads = 0
    tails = 0
    for _ in range(tosses):
        if random.random() < 0.5:
            tails += 1
        else:
            heads += 1
    return heads, tails


for tosses in (100, 1_000, 10_000):
    heads, tails = simulate_coin_tosses(tosses)
    proportion = heads / tosses
    print(f"tosses={tosses:>6} heads={heads:>6} tails={tails:>6} proportion={proportion:.4f}")
````

The proportion approaches $1/2$ as the number of tosses grows, but the absolute difference between heads and tails does not shrink. Counts fluctuate by roughly $\sqrt{N}$; proportions fluctuate by roughly $1/\sqrt{N}$.

## 10. Common mistakes

| Mistake | Consequence | Correction |
| :--- | :--- | :--- |
| `range(1, 5)` assumed to include `5` | One missing iteration | Remember that `stop` is exclusive. |
| `range(len(items) + 1)` | `IndexError` | Use `range(len(items))`. |
| Appending to a list while iterating it | Skipped or repeated elements | Iterate over a copy or build a new list. |
| Relying on the loop variable after an empty loop | `NameError` | Initialise the name before the loop. |
| `break` inside nested loops | Only the inner loop exits | Return from a function or restructure the loops. |
