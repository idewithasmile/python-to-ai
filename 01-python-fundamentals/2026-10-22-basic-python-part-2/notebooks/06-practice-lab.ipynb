## 1. Warm-up: FizzBuzz

Print `Fizz` for multiples of `3`, `Buzz` for multiples of `5`, and `FizzBuzz` for multiples of both. Return the results as a list.

```python
def fizzbuzz(limit):
    results = []
    for number in range(1, limit + 1):
        if number % 15 == 0:
            results.append("FizzBuzz")
        elif number % 3 == 0:
            results.append("Fizz")
        elif number % 5 == 0:
            results.append("Buzz")
        else:
            results.append(str(number))
    return results


print(fizzbuzz(15))
assert fizzbuzz(5) == ["1", "2", "Fizz", "4", "Buzz"]
assert fizzbuzz(15)[-1] == "FizzBuzz"
```

## 2. Sum of squares

Compare a loop with the closed form

$$
\sum_{k=1}^{n} k^2 = \frac{n(n+1)(2n+1)}{6}.
$$

```python
def sum_of_squares(n):
    total = 0
    for k in range(1, n + 1):
        total += k * k
    return total


for n in range(1, 20):
    closed_form = n * (n + 1) * (2 * n + 1) // 6
    assert sum_of_squares(n) == closed_form

print(sum_of_squares(100))
```

## 3. Factorial and Euler's number

Implement both iteratively. Compare the approximation with `math.e`.

```python
import math


def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


def estimate_e(terms):
    return sum(1.0 / factorial(k) for k in range(terms))


assert factorial(0) == 1
assert factorial(5) == 120
assert abs(estimate_e(15) - math.e) < 1e-12
print(f"e ≈ {estimate_e(15):.15f}")
```

## 4. Pi with two series

Implement Gregory–Leibniz and Nilakantha. Return the estimate, not the error.

```python
def estimate_pi_gregory(terms):
    total = 0.0
    for k in range(1, terms + 1):
        total += (-1) ** (k + 1) / (2 * k - 1)
    return 4.0 * total


def estimate_pi_nilakantha(terms):
    total = 3.0
    for k in range(terms):
        denominator = (2 * k + 2) * (2 * k + 3) * (2 * k + 4)
        total += 4.0 * (-1) ** k / denominator
    return total


assert abs(estimate_pi_nilakantha(30) - math.pi) < 1e-5
assert abs(estimate_pi_gregory(10_000) - math.pi) < 1e-4
print(estimate_pi_nilakantha(20))
```

## 5. Newton's method

Return the estimate and the number of iterations. Raise `ValueError` for negative input.

```python
def sqrt_newton(number, tolerance=1e-12, max_iterations=100):
    if number < 0:
        raise ValueError("number must be non-negative")
    if number == 0:
        return 0.0, 0

    estimate = number / 2.0
    for iteration in range(1, max_iterations + 1):
        next_estimate = (estimate + number / estimate) / 2.0
        if abs(next_estimate - estimate) < tolerance:
            return next_estimate, iteration
        estimate = next_estimate

    raise RuntimeError("Newton's method did not converge")


root, iterations = sqrt_newton(2)
print(f"sqrt(2) ≈ {root:.15f} after {iterations} iterations")
assert abs(root - math.sqrt(2)) < 1e-12
```

## 6. String utilities

Implement vowel counting, reversal without slicing, and a palindrome check that ignores case and non-alphanumeric characters.

```python
def count_vowels(text):
    vowels = set("aeiou")
    return sum(1 for character in text.lower() if character in vowels)


def reverse_text(text):
    characters = []
    for index in range(len(text) - 1, -1, -1):
        characters.append(text[index])
    return "".join(characters)


def is_palindrome(text):
    cleaned = "".join(character.lower() for character in text if character.isalnum())
    return cleaned == cleaned[::-1]


assert count_vowels("Programming") == 3
assert reverse_text("Python") == "nohtyP"
assert is_palindrome("A man, a plan, a canal: Panama")
assert not is_palindrome("Python")
print("string utilities passed")
```

## 7. Parsing records

Parse a three-field CSV line into a dictionary. Raise `ValueError` when the field count is wrong, and `TypeError` when the input is not a string.

```python
def parse_record(record):
    if not isinstance(record, str):
        raise TypeError("record must be a string")
    fields = record.split(",")
    if len(fields) != 3:
        raise ValueError("record must have exactly three fields")
    return dict(zip(("student_id", "name", "date_of_birth"), fields, strict=True))


parsed = parse_record("001,ada,1990-01-01")
print(parsed)
assert parsed["name"] == "ada"

try:
    parse_record("001,ada")
except ValueError as error:
    print(type(error).__name__, error)
```

## 8. Word frequencies

Count words in a text file, normalise case, strip common punctuation, and return the five most common words. Break ties alphabetically.

```python
from collections import Counter
from pathlib import Path


def word_frequencies(path):
    text = path.read_text(encoding="utf-8")
    counts = Counter()
    for word in text.lower().split():
        cleaned = word.strip('.,!?;:"')
        if cleaned:
            counts[cleaned] += 1
    return counts


def find_data_directory():
    for candidate in (Path.cwd(), Path.cwd().parent):
        data_directory = candidate / "data"
        if data_directory.is_dir():
            return data_directory
    raise FileNotFoundError("Could not find the session data directory")


data_directory = find_data_directory()
print(data_directory)
counts = word_frequencies(data_directory / "text" / "hello_world.txt")
for word, count in sorted(counts.items(), key=lambda item: (-item[1], item[0])):
    print(f"{word:>10}: {count}")
```

## 9. Pascal's triangle

Each entry is the sum of the two entries above it:

$$
C(n,k) = C(n-1,k-1) + C(n-1,k).
$$

```python
def pascal_triangle(rows):
    if rows < 0:
        raise ValueError("rows must be non-negative")
    triangle = []
    for row_index in range(rows):
        row = [1]
        previous = triangle[row_index - 1] if row_index > 0 else []
        for column in range(1, row_index + 1):
            left = previous[column - 1]
            right = previous[column] if column < len(previous) else 0
            row.append(left + right)
        triangle.append(row)
    return triangle


for row in pascal_triangle(6):
    print(row)
assert pascal_triangle(4)[-1] == [1, 3, 3, 1]
```

## 10. Fibonacci: iteration and recursion

The naive recursive implementation is exponential. The iterative implementation is linear. Both are correct.

```python
def fibonacci_iterative(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    previous, current = 0, 1
    for _ in range(n):
        previous, current = current, previous + current
    return previous


def fibonacci_recursive(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n < 2:
        return n
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)


for n in range(15):
    assert fibonacci_iterative(n) == fibonacci_recursive(n)

print(fibonacci_iterative(30))
```

## 11. Euclid's algorithm

$$
\gcd(a,b) = \gcd(b, a \bmod b),
$$

with $\gcd(a,0) = a$.

```python
def gcd(a, b):
    a, b = abs(a), abs(b)
    while b:
        a, b = b, a % b
    return a


assert gcd(48, 18) == 6
assert gcd(-48, 18) == 6
assert gcd(7, 0) == 7
print(gcd(1071, 462))
```

## 12. Collatz length

Return the number of terms in the Collatz sequence starting at `start`, including both endpoints.

```python
def collatz_length(start):
    if start < 1:
        raise ValueError("start must be a positive integer")
    length = 1
    value = start
    while value != 1:
        value = value // 2 if value % 2 == 0 else 3 * value + 1
        length += 1
    return length


assert collatz_length(1) == 1
assert collatz_length(6) == 9
print({start: collatz_length(start) for start in range(1, 11)})
```

## 13. Sieve of Eratosthenes

Mark multiples of each prime as composite. The inner loop starts at `number * number` because smaller multiples have already been marked.

```python
def sieve(limit):
    if limit < 2:
        return []
    is_prime = [True] * (limit + 1)
    is_prime[0] = False
    is_prime[1] = False
    for number in range(2, int(limit**0.5) + 1):
        if is_prime[number]:
            for multiple in range(number * number, limit + 1, number):
                is_prime[multiple] = False
    return [number for number, prime in enumerate(is_prime) if prime]


primes = sieve(100)
print(primes)
assert primes[:5] == [2, 3, 5, 7, 11]
assert len(primes) == 25
```

## 14. Reading the Iris data

Read the CSV file, compute the mean sepal length, and count the species. This is the same pipeline used by larger data tools, written with the standard library.

```python
import csv


def iris_summary(path):
    lengths = []
    species_counts = {}
    with path.open(encoding="utf-8", newline="") as handle:
        reader = csv.DictReader(handle)
        for row in reader:
            lengths.append(float(row["SepalLength"]))
            species = row["Species"]
            species_counts[species] = species_counts.get(species, 0) + 1
    return sum(lengths) / len(lengths), species_counts


mean_sepal_length, species_counts = iris_summary(data_directory / "iris.csv")
print(f"mean sepal length: {mean_sepal_length:.4f}")
print(species_counts)
```

## 15. Student report

Read `students.csv`, sort by name, and write a report to a temporary file. The report should include the number of students and one line per student.

```python
from tempfile import TemporaryDirectory


def load_students(path):
    students = []
    with path.open(encoding="utf-8", newline="") as handle:
        reader = csv.DictReader(handle)
        for line_number, row in enumerate(reader, start=2):
            identifier = row["student_id"].strip()
            name = row["name"].strip()
            date_of_birth = row["date_of_birth"].strip()
            if not identifier or not name or not date_of_birth:
                raise ValueError(f"missing field on line {line_number}")
            students.append(
                {
                    "student_id": identifier,
                    "name": name,
                    "date_of_birth": date_of_birth,
                }
            )
    return students


def build_report(students):
    lines = [f"student count: {len(students)}"]
    for student in sorted(students, key=lambda item: item["name"]):
        lines.append(f"{student['student_id']}: {student['name']}")
    return "\n".join(lines) + "\n"


students = load_students(data_directory / "text" / "students.csv")
report = build_report(students)
print(report)

with TemporaryDirectory() as directory:
    output = Path(directory) / "report.txt"
    output.write_text(report, encoding="utf-8")
    assert output.read_text(encoding="utf-8") == report
```
