# While Loops

A `while` loop repeats while a condition remains true. It is the right tool when the number of iterations is discovered during execution rather than known in advance.

## Learning goals

- Write a `while` loop with a clear initialisation, condition, body, and update.
- Use sentinels, `while True`, and `break` appropriately.
- State and check a loop invariant.
- Validate input without hiding control flow.
- Recognise the `continue` trap and other causes of infinite loops.

## 1. A counted loop

The update `i += 1` moves the state toward the exit condition. Without it, the loop never terminates.

```python
i = 0
while i < 5:
    print(i)
    i += 1

print("Done")
```

| Iteration | `i` before test | Test | Output | `i` after |
| :---: | :---: | :---: | :---: | :---: |
| 1 | 0 | `0 < 5` | `0` | 1 |
| 2 | 1 | `1 < 5` | `1` | 2 |
| 3 | 2 | `2 < 5` | `2` | 3 |
| 4 | 3 | `3 < 5` | `3` | 4 |
| 5 | 4 | `4 < 5` | `4` | 5 |
| 6 | 5 | `5 < 5` is false | none | 5 |

A countdown reverses the update.

```python
countdown = 5
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Liftoff")
```

## 2. Sentinel-controlled loops

A sentinel is a special value that signals the end of input. The loop reads until the sentinel appears. Always check the index before indexing the collection.

```python
def sum_until_zero(values):
    total = 0
    index = 0
    while index < len(values) and values[index] != 0:
        total += values[index]
        index += 1
    return total


print(sum_until_zero([4, 7, 2, 0, 99]))
print(sum_until_zero([1, 2, 3]))
```

The condition uses short-circuit evaluation. If `index < len(values)` is false, the second part is not evaluated, so the loop never indexes past the end.

## 3. `while True` with an explicit exit

An intentionally infinite loop with an obvious exit is often clearer than a compound condition.

```python
import random

random.seed(7)
while True:
    number = random.randint(1, 10)
    print(f"Drew {number}")
    if number == 7:
        break

print("Stopped after drawing a seven")
```

Every `while True` needs at least one reachable exit. If the exit depends on user input, validate that input and make every branch either continue or return.

## 4. Input validation

The function below asks until it receives a positive integer. It does not run in the notebook because it calls `input`, but the structure is the important part.

```python
def read_positive_integer(prompt):
    while True:
        raw = input(prompt)
        try:
            value = int(raw)
        except ValueError:
            print("Please enter a whole number.")
            continue
        if value <= 0:
            print("Please enter a positive number.")
            continue
        return value


# value = read_positive_integer("How many samples? ")
```

## 5. Loop invariants

An invariant is a statement that is true before the loop, remains true after each iteration, and helps establish the result when the loop ends.

```python
values = [3, 1, 4, 1, 5]
total = 0
index = 0

while index < len(values):
    # Invariant: total is the sum of values[0:index].
    total += values[index]
    index += 1

assert total == sum(values)
print(total)
```

## 6. `while` with `else`

The `else` block runs when the condition becomes false. A `break` skips it. This mirrors the behaviour of `for` loops.

```python
def find_first_even(values):
    index = 0
    while index < len(values):
        if values[index] % 2 == 0:
            print(f"Found {values[index]} at index {index}")
            break
        index += 1
    else:
        print("No even value found")


find_first_even([1, 3, 4, 5])
find_first_even([1, 3, 5])
```

## 7. Collatz sequences

The Collatz map is

$$
n \mapsto
\begin{cases}
n/2 & n \text{ even},\\
3n+1 & n \text{ odd}.
\end{cases}
$$

The sequence is conjectured to reach `1` for every positive starting value. The loop is a natural fit because the stopping time is not known in advance.

```python
def collatz_sequence(start):
    if start < 1:
        raise ValueError("start must be a positive integer")
    sequence = [start]
    value = start
    while value != 1:
        value = value // 2 if value % 2 == 0 else 3 * value + 1
        sequence.append(value)
    return sequence


for start in (1, 6, 7, 27):
    sequence = collatz_sequence(start)
    print(f"start={start:>2} length={len(sequence):>3} sequence={sequence[:12]}")
```

## 8. The `continue` trap

`continue` jumps to the next condition check. If the update appears after `continue`, it may never run.

```python
i = 0
while i < 5:
    i += 1
    if i == 2:
        continue
    print(i)
```

In the version above, the update happens before the `continue`, so the loop terminates. The unsafe version would place `i += 1` at the end of the body and then skip it whenever `i == 2`.

## 9. Common causes of infinite loops

1. The condition variable is never updated.
2. The update moves away from the exit condition.
3. `continue` skips the update.
4. The condition uses an exact floating-point comparison that never becomes true.
5. The loop waits for input or an event that never arrives.

When debugging a loop, print the state at the top of the body and add a temporary iteration cap. Remove the diagnostic after the bug is fixed.
