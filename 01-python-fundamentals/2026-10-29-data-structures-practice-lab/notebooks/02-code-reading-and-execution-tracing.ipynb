## 1. Loop Control Flow and String Tokenization

Let us analyze sequential execution across ranges with custom strides, string splits, and early termination flags.


```python
# Drill 1: Stride evaluation
# range(start, stop, step) generates 0, 100, 200, 300, 400 (stops before 500)
n = 1
for i in range(0, 500, 100):
    n = i
print("Final n:", n)
assert n == 400

# Drill 2: String split indexing
sentence = "Deep learning models require quality data"
tokens = sentence.split()
second_token = tokens[1]
print("Second token:", second_token)
assert second_token == "learning"

# Drill 3: Accumulation with early break
accumulator = 0
for step in range(5):  # 0, 1, 2, 3, 4
    accumulator += step
    # step=0 -> acc=0
    # step=1 -> acc=1
    # step=2 -> acc=3 -> accumulator > 0 and 3 % 3 == 0 -> break!
    if accumulator > 0 and accumulator % 3 == 0:
        break
print("Break accumulator:", accumulator)
assert accumulator == 3
```

## 2. String Slicing and Transformation Chaining

Negative slicing extracts elements relative to the end of a sequence (`seq[-k:]` extracts the final $k$ elements). Method chaining applies successive transformations.


```python
# Drill 4: Negative Slicing
part_a = "Computational "
part_b = "Linear Algebra and Statistics"
combined = part_a + part_b
tail_slice = combined[-10:]
print("Tail 10 characters:", repr(tail_slice))
assert tail_slice == "Statistics"

# Drill 5: Chained String Methods
raw_text = "@@gradient descent optimization method@@"
cleaned = raw_text.strip("@").capitalize()
print("Cleaned text:", repr(cleaned))
assert cleaned == "Gradient descent optimization method"

# Drill 6: In-Order Unique Character Vocabulary
source_phrase = "deep learning"
unique_ordered_chars: list[str] = []
for ch in source_phrase:
    if ch not in unique_ordered_chars:
        unique_ordered_chars.append(ch)
print("Unique ordered characters:", unique_ordered_chars)
assert "".join(unique_ordered_chars) == "dep larnig"
```

## 3. Matrix Transposition and Value Clamping

In scientific workflows, linear transformations and data normalization rely on nested index loops and numerical bounding functions.


```python
# Drill 7: Matrix Transposition
# Given matrix X of shape (rows, cols), construct X^T of shape (cols, rows)
X = [[1, 2, 3], [4, 5, 6]]
num_rows = len(X)
num_cols = len(X[0])

# Preallocate transposed matrix with zeros
transposed = [[0 for _ in range(num_rows)] for _ in range(num_cols)]

for r in range(num_rows):
    for c in range(num_cols):
        transposed[c][r] = X[r][c]

print("Original shape:", (num_rows, num_cols))
print("Transposed shape:", (len(transposed), len(transposed[0])))
print("Transposed matrix:", transposed)
assert transposed == [[1, 4], [2, 5], [3, 6]]


# Drill 8: Value Clamping (Min-Max Saturation)
def clamp_sequence(data: list[float], min_val: float, max_val: float) -> list[float]:
    """Clip all values in data to the closed interval [min_val, max_val]."""
    clamped = []
    for val in data:
        if val < min_val:
            clamped.append(min_val)
        elif val > max_val:
            clamped.append(max_val)
        else:
            clamped.append(val)
    return clamped


raw_readings = [10.5, 2.0, 5.4, -1.2, 0.5, 8.0]
bounded = clamp_sequence(raw_readings, min_val=0.0, max_val=5.0)
print("Bounded readings:", bounded)
assert bounded == [5.0, 2.0, 5.0, 0.0, 0.5, 5.0]
```

## 4. Tracing In-Place Bubble Sort

Bubble Sort iteratively compares adjacent elements and swaps them if out of order. With an early termination boolean flag (`swapped`), the algorithm halts as soon as the array is sorted.


```python
def bubble_sort_traced(arr: list[int]) -> tuple[list[int], int]:
    """Sort array in place while tracking adjacent swap operations."""
    items = arr[:]  # Work on copy for verification
    n = len(items)
    total_swaps = 0

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if items[j] > items[j + 1]:
                items[j], items[j + 1] = items[j + 1], items[j]
                total_swaps += 1
                swapped = True
        if not swapped:
            break

    return items, total_swaps


test_signal = [5, 1, 4, 2, 8]
sorted_signal, swaps = bubble_sort_traced(test_signal)
print(f"Sorted: {sorted_signal}, Total Swaps: {swaps}")
assert sorted_signal == [1, 2, 4, 5, 8]
assert swaps == 4
```
